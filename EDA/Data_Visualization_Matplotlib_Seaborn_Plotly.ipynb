## 1. What is data visualization?

Data visualization represents information using charts, graphs, maps, and plots. It helps us quickly identify:

- **Comparison:** Which product sold the most?
- **Trend:** Are sales increasing over time?
- **Distribution:** Where do most exam scores fall?
- **Relationship:** Does study time relate to marks?
- **Composition:** How is a total divided?
- **Patterns:** Are there clusters or unusual values?

| Question | Useful charts |
|---|---|
| Compare categories | Bar, column, dot plot |
| Show change over time | Line, area |
| Show distribution | Histogram, KDE, box, violin |
| Show relationships | Scatter, bubble, heatmap |
| Show composition | Stacked bar, pie |
| Show geographic patterns | Maps |

Choose a chart based on the question, not merely on appearance.

## 3. Bar chart and column chart

**Purpose:** Compare values across separate categories, such as sales by product or students by department. A vertical bar chart is often called a column chart. Use horizontal bars for long category names.

````python
counts = students["Department"].value_counts()
plt.figure(figsize=(7, 4))
plt.bar(counts.index, counts.values)
plt.title("Number of Students by Department")
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.show()
````

````python
plt.figure(figsize=(7, 4))
plt.barh(counts.index, counts.values)
plt.title("Students by Department")
plt.xlabel("Number of Students")
plt.ylabel("Department")
plt.show()
````

**Interpretation:** Bar height or length represents the value. Avoid 3D bars and unnecessary axis truncation.

## 5. Histogram

**Purpose:** Show the distribution of a numeric variable by dividing values into intervals called bins. Unlike a bar chart, the bars represent numeric ranges and usually touch.

````python
marks = [45, 50, 52, 55, 58, 60, 61, 62, 65, 66, 68, 70, 72,
         74, 75, 76, 78, 80, 82, 84, 85, 88, 90, 92, 95, 98]
plt.figure(figsize=(8, 4))
plt.hist(marks, bins=6, edgecolor="black")
plt.title("Distribution of Exam Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
````

````python

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

# First subplot: 4 bins
axes[0].hist(marks, bins=4, edgecolor="black")
axes[0].set_title("4 bins")

# Second subplot: 8 bins
axes[1].hist(marks, bins=8, edgecolor="black")
axes[1].set_title("8 bins")

# Third subplot: 15 bins
axes[2].hist(marks, bins=15, edgecolor="black")
axes[2].set_title("15 bins")

plt.tight_layout()
plt.show()
````

## 6. Pie chart and donut chart

**Purpose:** Show parts of a meaningful whole when there are only a few categories. Use a bar chart when accurate comparisons are more important.

````python
expenses = pd.Series(
    [40000, 25000, 15000, 10000, 10000],
    index=["Rent", "Food", "Transport", "Bills", "Other"]
)
plt.figure(figsize=(6, 6))
plt.pie(expenses.values, labels=expenses.index, autopct="%1.1f%%", startangle=90)
plt.title("Monthly Expense Distribution")
plt.show()
````

````python
plt.figure(figsize=(6, 6))
plt.pie(expenses.values, labels=expenses.index, autopct="%1.1f%%",
        wedgeprops={"width": 0.4}, startangle=90)
plt.title("Monthly Expenses — Donut Chart")
plt.show()
````

Do not use pie charts for negative values, unrelated quantities, or dozens of categories.

## 7. Scatter plot

**Purpose:** Explore the relationship between two numeric variables. Each dot represents one observation.

````python
plt.figure(figsize=(7, 5))
plt.scatter(students["StudyHours"], students["Marks"])
plt.title("Study Hours vs Exam Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
````

````python
plt.figure(figsize=(7, 5))
for dept in students["Department"].unique():
    part = students[students["Department"] == dept]
    plt.scatter(part["StudyHours"], part["Marks"], label=dept, s=80)
plt.title("Study Hours vs Marks by Department")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.legend(title="Department")
plt.show()
````

Look for positive, negative, or no apparent association; clusters; and unusual points. **Correlation does not prove causation.**

## 8. Box plot

**Purpose:** Summarize a numeric distribution using quartiles, median, whiskers, and potential outliers. Useful for comparing groups.

````python
plt.figure(figsize=(7, 4))
plt.boxplot(students["Marks"], tick_labels=["Marks"])
plt.title("Distribution of Marks")
plt.ylabel("Marks")
plt.show()
````

````python
sns.boxplot(data=students, x="Department", y="Marks")
plt.title("Marks Distribution by Department")
plt.show()
````

## 9. Violin plot

**Purpose:** Compare distributions while also showing a smoothed density shape. Wider sections indicate more observations around that value, based on the density estimate.

````python
sns.violinplot(data=students, x="Department", y="Marks", inner="quartile")
plt.title("Marks Distribution by Department")
plt.show()
````

Use violin plots when distribution shape matters and there are enough observations. With tiny samples, the smoothed shape can mislead.

## 10. Area chart

**Purpose:** Show magnitude or cumulative totals over an ordered axis, often time. Stacked areas can show how components contribute to a total.

````python
area_data = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Web": [20, 25, 30, 35, 40],
    "Mobile": [15, 20, 25, 30, 38],
    "Other": [5, 8, 7, 10, 12]
}).set_index("Month")
area_data.plot(kind="area", figsize=(8, 4), alpha=0.7)
plt.title("Traffic Sources Over Time")
plt.ylabel("Visitors (thousands)")
plt.show()
````

## 11. Grouped and stacked bar charts

**Grouped bars:** Compare subcategories side by side.  
**Stacked bars:** Show how subcategories contribute to each total.  
A 100% stacked bar compares proportions rather than absolute totals.

````python
comparison = pd.DataFrame({
    "Department": ["CS", "SE", "AI"],
    "Male": [30, 25, 20],
    "Female": [25, 35, 30]
}).set_index("Department")
comparison.plot(kind="bar", figsize=(7, 4), title="Grouped Bar Chart")
plt.ylabel("Students")
plt.show()
````

````python
comparison.plot(kind="bar", stacked=True, figsize=(7, 4), title="Stacked Bar Chart")
plt.ylabel("Students")
plt.show()
````

## 12. Heatmap

**Purpose:** Represent a matrix of values using color intensity. Common uses include correlation matrices, attendance schedules, and sales by month and region.

````python
corr = students[["StudyHours", "Marks", "Attendance"]].corr()
plt.figure(figsize=(6, 4))
plt.imshow(corr, aspect="auto", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")
plt.xticks(range(len(corr.columns)), corr.columns)
plt.yticks(range(len(corr.index)), corr.index)
plt.title("Correlation Heatmap")
plt.show()
````

````python
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Annotated Correlation Matrix")
plt.show()
````

## 13. KDE and density plot

**Purpose:** Show a smoothed estimate of a numeric distribution. KDE is related to a histogram but does not use visible bins.

````python
sns.kdeplot(data=students, x="Marks", fill=True)
plt.title("Density of Student Marks")
plt.show()
````

KDE depends on smoothing bandwidth and can be misleading for very small samples or bounded data.

## 14. Bubble chart

**Purpose:** A scatter plot where marker size represents a third numeric variable.

````python
bubble = pd.DataFrame({
    "Advertising": [10, 20, 30, 40, 50],
    "Revenue": [30, 45, 50, 70, 90],
    "Customers": [100, 180, 150, 260, 400]
})
plt.figure(figsize=(7, 5))
plt.scatter(bubble["Advertising"], bubble["Revenue"], s=bubble["Customers"], alpha=0.5)
plt.title("Advertising vs Revenue (bubble size = customers)")
plt.xlabel("Advertising Spend")
plt.ylabel("Revenue")
plt.show()
````

## 15. Error bar chart

**Purpose:** Show an estimate with uncertainty or variability, such as standard deviation, standard error, or a confidence interval. Always state what the error bars mean.

````python
groups = ["A", "B", "C"]
means = [70, 78, 85]
errors = [5, 4, 6]
plt.figure(figsize=(7, 4))
plt.errorbar(groups, means, yerr=errors, fmt="o", capsize=6)
plt.title("Average Score with Error Bars")
plt.xlabel("Group")
plt.ylabel("Average Score")
plt.show()
````

## 21. Radar / spider chart

**Purpose:** Compare a small number of profiles across shared metrics. It can be hard to compare exact values, so use carefully.

````python
categories = ["Speed", "Accuracy", "Teamwork", "Creativity", "Communication"]
scores = [8, 9, 7, 8, 6]
angles = np.linspace(0, 2*np.pi, len(categories), endpoint=False).tolist()
angles += angles[:1]
scores += scores[:1]
fig, ax = plt.subplots(figsize=(6, 6), subplot_kw={"polar": True})
ax.plot(angles, scores, marker="o")
ax.fill(angles, scores, alpha=0.15)
ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories)
ax.set_ylim(0, 10)
ax.set_title("Skill Profile")
plt.show()
````

## 22. Other specialized graphs

- **Waterfall:** Explains how sequential increases and decreases produce a final value.
- **Treemap:** Shows hierarchical parts of a whole using nested rectangles.
- **Sankey:** Shows flows between stages or categories.
- **Funnel:** Shows how many records remain at each process stage.
- **Gantt:** Shows task timelines and durations.
- **Candlestick:** Shows financial open, high, low, and close prices.
- **Choropleth:** Colors geographic regions by a value.
- **Network graph:** Shows entities and relationships.

Specialized tools may include Plotly, GeoPandas, Folium, NetworkX, and treemap packages.

## 23. What is Matplotlib?

Matplotlib is a foundational Python visualization library. It gives detailed control over chart elements.

Two important objects:

- **Axes:** an individual chart inside a figure.

Typical workflow: create figure and axes → plot data → add labels/title/legend → adjust layout → display or save.

## 30. Seaborn categorical plots

````python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
sns.barplot(data=students, x="Department", y="Marks", ax=axes[0, 0])
axes[0, 0].set_title("Average Marks")
sns.countplot(data=students, x="Department", ax=axes[0, 1])
axes[0, 1].set_title("Student Count")
sns.boxplot(data=students, x="Department", y="Marks", ax=axes[1, 0])
axes[1, 0].set_title("Box Plot")
sns.violinplot(data=students, x="Department", y="Marks", ax=axes[1, 1])
axes[1, 1].set_title("Violin Plot")
plt.tight_layout()
plt.show()
````

Important: Seaborn `barplot` estimates a statistic (mean by default) and may show an uncertainty interval. It does not simply count rows. Use `countplot` for counts.

## 31. Seaborn distribution plots

````python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
sns.histplot(data=students, x="Marks", bins=6, ax=axes[0])
axes[0].set_title("Histogram")
sns.kdeplot(data=students, x="Marks", fill=True, ax=axes[1])
axes[1].set_title("KDE")
sns.ecdfplot(data=students, x="Marks", ax=axes[2])
axes[2].set_title("Empirical CDF")
plt.tight_layout()
plt.show()
````

An ECDF shows the proportion of observations less than or equal to each value. It avoids histogram bin choices.

# Part E — Choosing and communicating charts

## 40. Chart selection guide

| Goal | Recommended chart | Alternative |
|---|---|---|
| Compare categories | Bar | Dot / lollipop |
| Compare values over time | Line | Area |
| Numeric distribution | Histogram | KDE / ECDF |
| Compare group distributions | Box | Violin |
| Two numeric variables | Scatter | Hexbin |
| Many numeric relationships | Pairplot | Correlation heatmap |
| Parts of a whole | Stacked bar | Pie |
| Composition across groups | Stacked / 100% stacked bar | Small multiples |
| Matrix patterns | Heatmap | Cluster map |
| Uncertainty | Error bars | Confidence interval plot |
| Geographic variation | Choropleth | Point map |
| Process stages | Funnel | Sankey |
| Task timelines | Gantt | Timeline |
| Financial prices | Candlestick | OHLC |

## 41. Common visualization mistakes

1. Choosing a chart that does not match the question.
2. Missing titles, axis labels, or units.
3. Truncating bar-chart axes in a misleading way.
4. Using too many colors or categories.
5. Ignoring overplotting when points overlap.
6. Hiding missing values or outliers without explanation.
7. Confusing correlation with causation.
8. Using error bars without explaining what they represent.
9. Adding decorative 3D effects that distort perception.
10. Showing a chart without explaining its main insight.

## 42. Accessibility and communication

- Use readable font sizes and sufficient contrast.
- Do not rely on color alone; add labels, markers, or line styles.
- Use descriptive titles and units.
- Keep legends readable and avoid covering data.
- Add a written description when publishing charts.
- Prefer simple charts over decorative effects.

## 44. Project 2: Business sales analysis

**Scenario:** A retail business wants to understand sales across regions, products, and months.

### Tasks
1. Calculate revenue for each transaction.
2. Compare total revenue by region with a bar chart.
3. Show monthly revenue as a line chart.
4. Show product contribution using a sorted bar chart.
5. Compare revenue distributions across regions using a box plot.

````python
business_sales = pd.DataFrame({
    "Date": pd.to_datetime([
        "2025-01-03", "2025-01-10", "2025-01-18", "2025-02-02",
        "2025-02-12", "2025-02-20", "2025-03-05", "2025-03-16",
        "2025-03-21", "2025-04-02", "2025-04-14", "2025-04-22"
    ]),
    "Region": ["North", "South", "East", "North", "West", "South",
               "East", "West", "North", "South", "East", "West"],
    "Product": ["Laptop", "Phone", "Tablet", "Phone", "Laptop", "Tablet",
                "Phone", "Tablet", "Laptop", "Laptop", "Tablet", "Phone"],
    "Quantity": [2, 5, 3, 4, 1, 6, 3, 4, 2, 1, 5, 4],
    "UnitPrice": [1500, 600, 400, 600, 1500, 400, 600, 400, 1500, 1500, 400, 600]
})
business_sales["Revenue"] = business_sales["Quantity"] * business_sales["UnitPrice"]
business_sales
````

### Example solution

````python
revenue_region = business_sales.groupby("Region")["Revenue"].sum().sort_values(ascending=False)
revenue_region.plot(kind="bar", title="Revenue by Region", ylabel="Revenue")
plt.show()

business_sales["Month"] = business_sales["Date"].dt.to_period("M").astype(str)
monthly_revenue = business_sales.groupby("Month")["Revenue"].sum()
monthly_revenue.plot(kind="line", marker="o", title="Monthly Revenue")
plt.ylabel("Revenue")
plt.show()

product_revenue = business_sales.groupby("Product")["Revenue"].sum().sort_values()
product_revenue.plot(kind="barh", title="Revenue by Product")
plt.xlabel("Revenue")
plt.show()

sns.boxplot(data=business_sales, x="Region", y="Revenue")
plt.title("Transaction Revenue by Region")
plt.show()

heat = business_sales.pivot_table(index="Region", columns="Product",
                                  values="Revenue", aggfunc="sum", fill_value=0)
sns.heatmap(heat, annot=True, fmt=".0f")
plt.title("Revenue by Region and Product")
plt.show()

px.bar(business_sales.groupby("Product", as_index=False)["Revenue"].sum(),
       x="Product", y="Revenue", title="Interactive Revenue by Product").show()
````
