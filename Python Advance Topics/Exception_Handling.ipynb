## 1. What is an error?

An error is a problem that prevents a program from working as intended.

Consider:

```python
print(10 / 0)
```

Python cannot divide a number by zero. It raises an exception called `ZeroDivisionError`.

### Two common categories

**Syntax errors:** Python cannot understand the program's structure.

```python
# Missing closing parenthesis — syntax error
# print("Hello"
```

**Exceptions:** The syntax is valid, but an error occurs while the program is running.

```python
print(10 / 0)  # ZeroDivisionError
```

Exception handling helps a program respond to runtime problems instead of terminating unexpectedly.

## 2. What happens without exception handling?

````python
number1 = 10
number2 = 0

result = number1 / number2
print("Result:", result)

print("Program completed")
````

Run the previous cell and observe the result. Python raises `ZeroDivisionError`, so the later print statement does not execute.

Let's handle the problem.

## 3. Basic `try` and `except`

- `try`: contains code that may raise an exception.
- `except`: contains code that handles an exception.

### Syntax

```python
try:
    # Code that may fail
except:
    # Code to handle the error
```

For teaching and real applications, prefer catching a specific exception instead of using a bare `except`.

````python
try:
    result = 10 / 0
    print(result)
except ZeroDivisionError:
    print("You cannot divide by zero.")

print("Program completed")
````

## 4. Catching a specific exception

A specific `except` block handles only the exception type named in it.

For example, `ZeroDivisionError` handles division by zero, but it does not handle an unrelated error such as `ValueError`.

````python
try:
    number = int("hello")
    print(number)
except ZeroDivisionError:
    print("Division by zero is not allowed.")

print("End of program")
````

The previous example raises `ValueError`. Since we only caught `ZeroDivisionError`, the exception is not handled and the cell stops with an error.

Now catch the correct exception.

````python
try:
    number = int("hello")
    print(number)
except ValueError:
    print("Please provide a valid integer.")

print("End of program")
````

## 6. `ZeroDivisionError`

````python
try:
    a = 20
    b = 0
    print(a / b)
except ZeroDivisionError:
    print("Error: denominator cannot be zero.")
````

## 7. `ValueError`

`ValueError` often occurs when the type is acceptable but the value is not.

For example, `int()` can convert a numeric string, but it cannot convert `"hello"`.

````python
try:
    age = int("twenty")
    print("Age:", age)
except ValueError:
    print("Error: enter age as a number.")
````

## 8. `TypeError`

`TypeError` commonly occurs when an operation is not supported for the supplied types.

````python
try:
    result = 10 + "20"
    print(result)
except TypeError:
    print("Error: cannot add an integer and a string directly.")
````

## 12. `FileNotFoundError`

This exception occurs when Python tries to open a file that cannot be found at the requested path.

````python
from pathlib import Path

file_path = Path("this_file_does_not_exist.txt")

try:
    content = file_path.read_text(encoding="utf-8")
    print(content)
except FileNotFoundError:
    print("Error: the requested file was not found.")
````

## 13. Catching multiple specific exceptions

A program can fail in more than one way. You can write multiple `except` blocks.

Python checks them in order and executes the first matching block.

````python
try:
    number1 = int(input("Enter first number: "))
    number2 = int(input("Enter second number: "))

    result = number1 / number2
    print("Result:", result)

except ValueError:
    print("Please enter numbers only.")

except ZeroDivisionError:
    print("The second number cannot be zero.")
````

### Important: order of exception handlers

Catch more specific exceptions before more general exceptions.

For example, `FileNotFoundError` is a subclass of `OSError`. If you catch `OSError` first, a later `FileNotFoundError` handler will never be reached for that error.

````python
from pathlib import Path

try:
    Path("missing_file.txt").read_text(encoding="utf-8")

except FileNotFoundError:
    print("Specific error: file does not exist.")

except OSError:
    print("General file-system error.")
````

## 14. Catch multiple exception types in one block

If you want to handle different exception types in exactly the same way, use a tuple.

````python
try:
    value = int("abc")
    result = 10 / value

except (ValueError, ZeroDivisionError):
    print("The input or calculation is invalid.")
````

## 15. Get the actual exception message with `as`

You can capture the exception object using `as`.

This is useful for debugging and logging.

````python
try:
    result = 10 / 0

except ZeroDivisionError as error:
    print("Exception type:", type(error).__name__)
    print("Exception message:", error)
````

## 16. The `else` block

The `else` block runs only when the `try` block completes without raising an exception.

Structure:

```python
try:
    # Risky code
except SomeException:
    # Error handling
else:
    # Runs if no exception occurred
```

````python
try:
    number = int("100")

except ValueError:
    print("Invalid number.")

else:
    print("Conversion successful.")
    print("Number:", number)
````

### Try the same example with invalid input

````python
try:
    number = int("abc")

except ValueError:
    print("Invalid number.")

else:
    print("Conversion successful.")
    print("Number:", number)
````

## 17. The `finally` block

The `finally` block runs whether an exception occurs or not, including when a handled exception is raised.

It is useful for cleanup operations, such as closing resources. For files, `with open(...)` is usually the preferred way to manage file closure.

````python
try:
    print("Opening a resource...")
    result = 10 / 0

except ZeroDivisionError:
    print("An error occurred.")

finally:
    print("Cleanup operation completed.")
````

### `finally` also runs when there is no exception

````python
try:
    result = 10 / 2
    print("Result:", result)

except ZeroDivisionError:
    print("Cannot divide by zero.")

finally:
    print("This block always runs.")
````

## 18. Understanding the complete structure

A `try` statement can use `except`, `else`, and `finally`.

The order is:

1. `try`
2. One or more `except` blocks
3. Optional `else`
4. Optional `finally`

````python
try:
    number = int("25")
    result = 100 / number

except ValueError:
    print("Invalid input.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Operation successful.")
    print("Result:", result)

finally:
    print("Program section finished.")
````

## 19. Raising an exception using `raise`

Sometimes Python does not automatically know that a value is invalid according to your application's rules.

Use `raise` to deliberately raise an exception.

Example: a negative age is not valid for a simple age-registration function.

````python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative.")
````

### Handle your raised exception

````python
try:
    age = -5

    if age < 0:
        raise ValueError("Age cannot be negative.")

    print("Age:", age)

except ValueError as error:
    print("Invalid age:", error)
````

## 20. Validate user input with a function

Let's create a function that accepts an age and raises a specific exception if the age is invalid.

````python
def register_student(name, age):
    if not name:
        raise ValueError("Name cannot be empty.")

    if age < 0:
        raise ValueError("Age cannot be negative.")

    print("Student registered:", name, age)


try:
    register_student("Ali", -2)

except ValueError as error:
    print("Registration failed:", error)
````

## 21. Re-raising an exception

Sometimes you want to catch an exception temporarily, perform an action such as logging, and then let the original exception continue to the caller.

Use `raise` without an argument inside an `except` block to re-raise the current exception.

````python
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Logging: division by zero was attempted.")
        raise


try:
    divide(10, 0)
except ZeroDivisionError:
    print("The caller handled the exception.")
````

## 22. Custom exceptions

Python allows you to create your own exception classes.

A custom exception should usually inherit from `Exception`.

Example: a bank application can define an `InsufficientBalanceError`.

````python
class InsufficientBalanceError(Exception):
    pass


def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError("Not enough balance.")

    return balance - amount


try:
    new_balance = withdraw(5000, 7000)
    print("Remaining balance:", new_balance)

except InsufficientBalanceError as error:
    print("Withdrawal failed:", error)
````

## 23. Custom exception with additional information

````python
class InvalidMarksError(Exception):
    def __init__(self, marks):
        self.marks = marks
        super().__init__(f"Marks must be between 0 and 100. Received: {marks}")


def assign_grade(marks):
    if marks < 0 or marks > 100:
        raise InvalidMarksError(marks)

    if marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    else:
        return "Needs improvement"


try:
    print(assign_grade(120))
except InvalidMarksError as error:
    print(error)
    print("Invalid value:", error.marks)
````

## 24. Exception handling with files

File operations can fail for several reasons:

- The file does not exist.
- Access is denied.
- The file cannot be decoded using the selected encoding.
- Another operating-system error occurs.

Use a context manager (`with`) to manage the file safely.

````python
from pathlib import Path

file_path = Path("exception_practice.txt")

try:
    with file_path.open("r", encoding="utf-8") as file:
        content = file.read()

except FileNotFoundError:
    print("File not found.")

except PermissionError:
    print("Permission denied.")

except UnicodeDecodeError:
    print("The file could not be decoded as UTF-8.")

except OSError as error:
    print("File-system error:", error)
````

### Create a file, then read it safely

````python
file_path.write_text("Python exception handling", encoding="utf-8")

try:
    with file_path.open("r", encoding="utf-8") as file:
        print(file.read())

except OSError as error:
    print("Could not read file:", error)
````

## 25. Exception handling with loops

A common requirement is to keep asking the user for input until valid data is provided.

The following example repeats until the user enters a valid integer.

````python
while True:
    try:
        age = int(input("Enter your age: "))
        if age < 0:
            print("Age cannot be negative.")
            continue

        print("Your age is:", age)
        break

    except ValueError:
        print("Invalid input. Please enter a whole number.")
````

## 26. A practical calculator with specific exception handling

This example catches invalid numeric input and division by zero separately.

````python
try:
    first_number = float(input("Enter first number: "))
    second_number = float(input("Enter second number: "))
    operator = input("Enter operator (+, -, *, /): ")

    if operator == "+":
        answer = first_number + second_number
    elif operator == "-":
        answer = first_number - second_number
    elif operator == "*":
        answer = first_number * second_number
    elif operator == "/":
        answer = first_number / second_number
    else:
        raise ValueError("Unsupported operator.")

except ValueError as error:
    print("Input error:", error)

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Answer:", answer)

finally:
    print("Calculator operation finished.")
````

## 27. Common mistakes in exception handling

### Mistake 1: Catching every exception with a bare `except`

```python
# Avoid this in normal application code:
try:
    result = 10 / 0
except:
    print("Something went wrong")
```

Why is it a problem?

- It hides which problem occurred.
- It can accidentally catch exceptions you did not intend to handle.
- It makes debugging harder.

Prefer:

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
```

### Mistake 2: Catching the wrong exception

If `int("hello")` raises `ValueError`, catching only `ZeroDivisionError` will not handle it.

### Mistake 3: Putting too much code inside `try`

Keep the `try` block focused on the operation that may fail. This makes it easier to identify the source of an exception.

### Mistake 4: Silently ignoring errors

Avoid `except SomeError: pass` unless there is a deliberate, documented reason to ignore that specific failure.

## 28. Exception hierarchy

Built-in exceptions follow an inheritance hierarchy.

A simplified view:

```text
BaseException
    |
    +-- Exception
          |
          +-- ArithmeticError
          |      |
          |      +-- ZeroDivisionError
          |      +-- OverflowError
          |
          +-- LookupError
          |      |
          |      +-- IndexError
          |      +-- KeyError
          |
          +-- ValueError
          +-- TypeError
          +-- OSError
                 |
                 +-- FileNotFoundError
                 +-- PermissionError
```

This explains why a general handler can catch more specific exceptions. In application code, catch the narrowest useful exception type.

````python
print(issubclass(ZeroDivisionError, ArithmeticError))
print(issubclass(FileNotFoundError, OSError))
print(issubclass(ValueError, Exception))
````

## 35. Solution 2 — Student marks validation

````python
def get_grade(marks):
    if not isinstance(marks, (int, float)):
        raise TypeError("Marks must be numeric.")

    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100.")

    if marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    else:
        return "F"


try:
    print("Grade:", get_grade(85))
except (TypeError, ValueError) as error:
    print("Invalid marks:", error)
````

## 38. Solution 5 — Bank withdrawal

````python
class InsufficientBalanceError(Exception):
    pass


class BankAccount:
    def __init__(self, holder, balance):
        self.holder = holder
        self.balance = balance

    def withdraw(self, amount):
        if amount < 0:
            raise ValueError("Withdrawal amount cannot be negative.")

        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient account balance.")

        self.balance -= amount
        return self.balance


account = BankAccount("Ali", 10000)

try:
    remaining = account.withdraw(12000)
    print("Remaining balance:", remaining)

except ValueError as error:
    print("Invalid amount:", error)

except InsufficientBalanceError as error:
    print("Transaction failed:", error)
````
