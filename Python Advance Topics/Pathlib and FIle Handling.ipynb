## 1. What is `pathlib`?

A **path** is the address of a file or folder on your computer.

Examples:

- Windows: `C:\Users\Student\Documents\notes.txt`
- macOS/Linux: `/home/student/Documents/notes.txt`

Python's built-in `pathlib` module provides an object-oriented way to work with paths.

Instead of manually joining strings such as:

```python
folder + "/" + filename
```

you can use:

```python
folder / filename
```

The `/` operator joins path parts. It does not divide numbers here.

`pathlib` is included with Python, so you do not need to install it.

### Path parts

A path can contain a folder name, a file name, and a file extension.

````python
path = Path("students/data/marks.csv")

print("Full path:", path)
print("Name:", path.name)
print("Stem (name without extension):", path.stem)
print("Suffix (extension):", path.suffix)
print("Parent folder:", path.parent)
````

## 5. Joining paths

Use `/` to combine path parts. `pathlib` automatically uses the correct path separator for the operating system.

````python
base_folder = Path("students")
data_folder = base_folder / "data"
marks_file = data_folder / "marks.csv"

print(base_folder)
print(data_folder)
print(marks_file)
````

### Build a path from multiple parts

You can also use `joinpath()`.

````python
path1 = Path("students").joinpath("data", "marks.csv")
print(path1)
````

## 6. Absolute paths and relative paths

- A **relative path** is interpreted from the current working directory. Example: `data/marks.csv`
- An **absolute path** describes a complete location from the file-system root.

Use `resolve()` to obtain an absolute, normalized path.

````python
relative_path = Path("data/marks.csv")

print("Relative path:", relative_path)
print("Absolute path:", relative_path.resolve())
````

## 7. Create a safe practice workspace

We will keep all file-creation and deletion exercises inside one practice directory.

`mkdir()` creates a directory.

- `parents=True` creates missing parent directories.
- `exist_ok=True` prevents an error if the directory already exists.

````python
workspace = Path.cwd() / "pathlib_practice"
workspace.mkdir(exist_ok=True)

print("Practice folder:", workspace.resolve())
print("Folder exists:", workspace.exists())
print("Is a directory:", workspace.is_dir())
````

## 8. Create subfolders

## 9. Create files

`Path.touch()` creates an empty file if it does not exist. If it already exists, it updates its modification time.

````python
empty_file = documents_folder / "empty.txt"
empty_file.touch()

print("File exists:", empty_file.exists())
print("Is a file:", empty_file.is_file())
````

## 10. Writing text to a file

`write_text()` writes text to a file. If the file already exists, its previous content is replaced.

The `encoding="utf-8"` argument is a good default for text files.

````python
student_file = documents_folder / "student.txt"

student_file.write_text(
    "Name: Ali\nDepartment: Computer Science\nSemester: 5",
    encoding="utf-8"
)

print("Text written successfully.")
````

## 11. Reading text from a file

`read_text()` reads the complete text content of a file.

````python
content = student_file.read_text(encoding="utf-8")

print(content)
````

### Read a file line by line

Use `open()` with a `Path` object when you want to process lines individually.

````python
with student_file.open("r", encoding="utf-8") as file:
    for line_number, line in enumerate(file, start=1):
        print(line_number, line.strip())
````

## 12. Append content to a file

Use `open("a")` to append content without removing the existing content.

````python
with student_file.open("a", encoding="utf-8") as file:
    file.write("\nStatus: Active")

print(student_file.read_text(encoding="utf-8"))
````

## 13. Writing and reading CSV-style text

`pathlib` handles paths; Python's `csv` module handles CSV data. They work together.

````python
import csv

students_csv = documents_folder / "students.csv"

student_rows = [
    ["id", "name", "department"],
    [1, "Ali", "Computer Science"],
    [2, "Sara", "Software Engineering"],
    [3, "Usman", "Data Science"],
]

with students_csv.open("w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(student_rows)

with students_csv.open("r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
````

## 14. Check whether a path exists

Common checks:

- `exists()` — does the path exist?
- `is_file()` — is it a file?
- `is_dir()` — is it a directory?

````python
paths_to_check = [
    workspace,
    student_file,
    documents_folder / "missing.txt"
]

for path in paths_to_check:
    print(
        path.name,
        "| exists:", path.exists(),
        "| file:", path.is_file(),
        "| directory:", path.is_dir()
    )
````

## 15. List files and folders

`iterdir()` returns the direct contents of a directory. It does not automatically search inside subdirectories.

````python
print("Items inside the practice folder:")

for item in workspace.iterdir():
    kind = "Folder" if item.is_dir() else "File"
    print(kind, "-", item.name)
````

### List only files or only folders

````python
all_files = [item for item in workspace.iterdir() if item.is_file()]
all_folders = [item for item in workspace.iterdir() if item.is_dir()]

print("Files:")
for item in all_files:
    print(item.name)

print("\nFolders:")
for item in all_folders:
    print(item.name)
````

## 16. Search using `glob()`

`glob()` searches for paths matching a pattern.

Examples:

- `*.txt` — all text files in this folder
- `*.csv` — all CSV files in this folder
- `student*` — names beginning with `student`

````python
print("Text files in documents:")
for path in documents_folder.glob("*.txt"):
    print(path.name)

print("\nCSV files in documents:")
for path in documents_folder.glob("*.csv"):
    print(path.name)
````

### Search recursively using `rglob()`

`rglob()` searches in the folder and all its subfolders.

````python
print("All files under the workspace:")
for path in workspace.rglob("*"):
    if path.is_file():
        print(path.relative_to(workspace))
````

## 17. Rename a file

`rename()` changes a file's name or location. The destination should be chosen carefully because an existing destination may be replaced depending on the operating system.

````python
old_path = documents_folder / "student.txt"
new_path = documents_folder / "student_profile.txt"

if old_path.exists() and not new_path.exists():
    old_path.rename(new_path)

print("New file exists:", new_path.exists())
````

## 18. Move a file to another folder

A move can be performed by renaming a path to a destination in another folder.

````python
source_file = documents_folder / "empty.txt"
destination_file = backup_folder / "empty_backup.txt"

if source_file.exists() and not destination_file.exists():
    source_file.rename(destination_file)

print("Moved file exists:", destination_file.exists())
````

## 19. Copying files

`pathlib` focuses on paths and does not provide a dedicated `copy()` method. Use Python's `shutil` module for copying.

````python
import shutil

copy_source = students_csv
copy_destination = backup_folder / "students_copy.csv"

shutil.copy2(copy_source, copy_destination)

print("Copy created:", copy_destination.exists())
````

## 20. Delete files and folders

Be careful with deletion. These operations cannot always be undone.

- `unlink()` deletes a file.
- `rmdir()` deletes an empty directory.
- `shutil.rmtree()` deletes a directory and everything inside it.

The following examples delete only specific files created in this practice workspace.

````python
temporary_file = workspace / "temporary.txt"
temporary_file.write_text("This file will be deleted.", encoding="utf-8")

if temporary_file.exists() and temporary_file.is_file():
    temporary_file.unlink()

print("Temporary file exists:", temporary_file.exists())
````

### Remove an empty folder

````python
empty_folder = workspace / "empty_folder"
empty_folder.mkdir(exist_ok=True)

if empty_folder.exists() and not any(empty_folder.iterdir()):
    empty_folder.rmdir()

print("Empty folder exists:", empty_folder.exists())
````

## 21. File information and metadata

Useful properties and methods:

- `stat().st_size` — file size in bytes
- `stat().st_mtime` — last modification time as a timestamp
- `suffix` — file extension
- `name` — file name

````python
for path in workspace.rglob("*"):
    if path.is_file():
        info = path.stat()
        print("File:", path.name)
        print("Size:", info.st_size, "bytes")
        print("Extension:", path.suffix)
        print("Modified timestamp:", info.st_mtime)
        print("-" * 30)
````

## 22. File extensions and suffixes

A suffix is the extension at the end of a file name. `suffixes` returns all suffixes, which is useful for names such as `archive.tar.gz`.

````python
examples = [
    Path("report.pdf"),
    Path("photo.png"),
    Path("archive.tar.gz"),
    Path("README")
]

for path in examples:
    print(path.name, "=> suffix:", path.suffix, "| all suffixes:", path.suffixes)
````

## 23. Change a file extension

`with_suffix()` returns a new path with a different extension. It does not rename the actual file by itself.

````python
pdf_path = Path("report.pdf")
text_path = pdf_path.with_suffix(".txt")

print("Original path:", pdf_path)
print("New path:", text_path)
````

## 24. Get a path relative to another path

`relative_to()` expresses one path in relation to a parent path. The target must be inside the parent path.

````python
example_file = workspace / "documents" / "students.csv"

print("Full path:", example_file)
print("Relative to workspace:", example_file.relative_to(workspace))
````

## 25. Parent, parents, and name

- `parent` gives the immediate parent folder.
- `parents` provides all parent folders.
- `name` gives the final path component.

````python
sample_path = Path("university/semester5/python/assignment.py")

print("Name:", sample_path.name)
print("Parent:", sample_path.parent)

print("All parents:")
for parent in sample_path.parents:
    print(parent)
````

## 26. Safe file handling with exceptions

A file may not exist, or your program may not have permission to access it. Use `try` and `except` to handle expected errors.

````python
missing_file = workspace / "does_not_exist.txt"

try:
    text = missing_file.read_text(encoding="utf-8")
    print(text)
except FileNotFoundError:
    print("The requested file does not exist.")
except PermissionError:
    print("You do not have permission to read this file.")
````

## 27. Check before creating or overwriting

Use `exists()` to decide what action to take. This example avoids overwriting an existing report.

````python
report_file = reports_folder / "summary.txt"

if report_file.exists():
    print("Report already exists. No changes made.")
else:
    report_file.write_text("Weekly summary", encoding="utf-8")
    print("Report created.")
````

## 28. Common path operations — quick reference

| Task | Method / expression |

| File name | `path.name` |
| Name without extension | `path.stem` |
| Extension | `path.suffix` |
| Parent folder | `path.parent` |
| Check existence | `path.exists()` |
| Check file | `path.is_file()` |
| Check directory | `path.is_dir()` |
| Create folder | `path.mkdir()` |
| Create empty file | `path.touch()` |
| Read text | `path.read_text()` |
| Write text | `path.write_text()` |
| Delete file | `path.unlink()` |
| List direct contents | `path.iterdir()` |
| Search pattern | `path.glob("*.txt")` |
| Recursive search | `path.rglob("*.txt")` |
| Rename / move | `path.rename(destination)` |

Remember: copying is usually done with `shutil.copy()` or `shutil.copy2()`.

### Solution 5 — File organizer

````python
downloads = workspace / "downloads"
images_folder = downloads / "images"
documents_downloads = downloads / "documents"
other_folder = downloads / "other"

for folder in [images_folder, documents_downloads, other_folder]:
    folder.mkdir(parents=True, exist_ok=True)

sample_downloads = {
    "photo1.png": "Image one",
    "photo2.jpg": "Image two",
    "report.pdf": "PDF placeholder",
    "notes.txt": "Text placeholder"
}

for filename, content in sample_downloads.items():
    (downloads / filename).write_text(content, encoding="utf-8")

for file_path in list(downloads.iterdir()):
    if not file_path.is_file():
        continue

    if file_path.suffix.lower() in [".png", ".jpg", ".jpeg", ".gif"]:
        destination_folder = images_folder
    elif file_path.suffix.lower() in [".pdf", ".txt", ".docx"]:
        destination_folder = documents_downloads
    else:
        destination_folder = other_folder

    file_path.rename(destination_folder / file_path.name)

print("Organizer completed.")
````

### Solution 6 — Backup a file

````python
important_file = workspace / "important.txt"
backup_file = backup_folder / "important.txt"

important_file.write_text("This is important information.", encoding="utf-8")
shutil.copy2(important_file, backup_file)

original_content = important_file.read_text(encoding="utf-8")
backup_content = backup_file.read_text(encoding="utf-8")

print("Original:", original_content)
print("Backup:", backup_content)
print("Contents match:", original_content == backup_content)
````
