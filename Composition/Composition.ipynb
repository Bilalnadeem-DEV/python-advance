## 1. Quick OOP revision

A **class** is a blueprint. An **object** is an instance created from that blueprint.

For example, `Car` is a class, while `my_car` is an object of that class.

````python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def drive(self):
        print(self.brand, "is driving")


my_car = Car("Toyota", "White")

print(my_car.brand)
print(my_car.color)
my_car.drive()
````

## 2. What is inheritance?

**Inheritance** allows one class to reuse and extend the attributes and methods of another class.

- Parent class: the class whose features are inherited.
- Child class: the class that inherits those features.

Think of this relationship:

- A `Dog` **is an** `Animal`.
- A `Car` **is a** `Vehicle`.
- A `Teacher` **is a** `Person`.

This is called an **is-a relationship**.

### Basic syntax

```python
class Parent:
    pass

class Child(Parent):
    pass
```

## 3. First inheritance example: Animal and Dog

````python
class Animal:
    def eat(self):
        print("This animal is eating")


class Dog(Animal):
    def bark(self):
        print("The dog is barking")


my_dog = Dog()

my_dog.eat()   # Inherited from Animal
my_dog.bark()  # Defined in Dog
````

### Understand what happened

`Dog` does not define an `eat()` method. It inherits that method from `Animal`.

The object `my_dog` can therefore use both:

- `eat()` — inherited from the parent.
- `bark()` — defined in the child.

This is code reuse through inheritance.

## 4. Inheritance with constructors (`__init__`)

A constructor initializes an object when it is created.

When a child class defines its own `__init__()`, it does not automatically execute the parent class's constructor. To initialize the parent part, use `super()`.

````python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, roll_number):
        super().__init__(name)
        self.roll_number = roll_number


student = Student("Ayesha", "CS-101")

print(student.name)
print(student.roll_number)
````

### What does `super()` do?

`super()` gives access to the next class in the method resolution order (MRO).

In this example:

```python
super().__init__(name)
```

calls the parent class's constructor so that `self.name` is initialized.

Without this call, `name` would not be initialized by `Person` in this example.

## 5. Method overriding

A child class can provide its own implementation of a method already defined in its parent class. This is called **method overriding**.

````python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog says: Woof!")


class Cat(Animal):
    def sound(self):
        print("Cat says: Meow!")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
````

### Calling the parent method with `super()`

Sometimes the child wants to extend the parent's behavior rather than completely replace it.

````python
class Employee:
    def show_details(self):
        print("Employee details")


class Manager(Employee):
    def show_details(self):
        super().show_details()
        print("Manager has team responsibilities")


manager = Manager()
manager.show_details()
````

## 6. Types of inheritance

Python supports several inheritance structures.

### A. Single inheritance

One child inherits from one parent.

```text
Animal
   |
  Dog
```

````python
class Animal:
    def eat(self):
        print("Eating")


class Dog(Animal):
    def bark(self):
        print("Barking")


dog = Dog()
dog.eat()
dog.bark()
````

### B. Multilevel inheritance

A class inherits from a child class, creating a chain.

```text
Animal
   |
 Mammal
   |
  Dog
```

````python
class Animal:
    def eat(self):
        print("Eating")


class Mammal(Animal):
    def walk(self):
        print("Walking")


class Dog(Mammal):
    def bark(self):
        print("Barking")


dog = Dog()
dog.eat()
dog.walk()
dog.bark()
````

### C. Hierarchical inheritance

Multiple child classes inherit from the same parent.

```text
       Animal
       /    \
     Dog    Cat
```

````python
class Animal:
    def eat(self):
        print("Eating")


class Dog(Animal):
    def bark(self):
        print("Woof!")


class Cat(Animal):
    def meow(self):
        print("Meow!")


dog = Dog()
cat = Cat()

dog.eat()
dog.bark()

cat.eat()
cat.meow()
````

### D. Multiple inheritance

One child class inherits from more than one parent.

```text
Father     Mother
    \       /
      Child
```

````python
class Camera:
    def take_photo(self):
        print("Photo captured")


class Phone:
    def make_call(self):
        print("Calling...")


class SmartPhone(Camera, Phone):
    pass


mobile = SmartPhone()
mobile.take_photo()
mobile.make_call()
````

### Method Resolution Order (MRO)

When multiple parent classes contain methods with the same name, Python follows the Method Resolution Order to decide which method is used.

Use `ClassName.mro()` to inspect the order.

````python
print(SmartPhone.mro())
````

## 7. Advantages and disadvantages of inheritance

### Advantages

- Reuses existing code.
- Represents natural *is-a* relationships.
- Supports method overriding and polymorphism.
- Can provide a consistent interface across related classes.

### Disadvantages

- Child classes become dependent on parent-class behavior.
- Deep inheritance chains can become difficult to understand.
- Changes in a parent class can affect many child classes.
- Inheritance may be inappropriate when the relationship is not truly *is-a*.

# Part 2 — Composition

## 8. What is composition?

**Composition** is an OOP technique in which one class contains or uses an object of another class.

Think of these relationships:

- A `Car` **has an** `Engine`.
- A `Computer` **has a** `Processor`.
- A `House` **has a** `Room`.
- A `ShoppingCart` **has** products.

This is called a **has-a relationship**.

Instead of inheriting another class's behavior, a class creates or receives an object and uses that object's methods.

## 9. First composition example: Car and Engine

````python
class Engine:
    def start(self):
        print("Engine started")


class Car:
    def __init__(self):
        self.engine = Engine()  # Car has an Engine object

    def start_car(self):
        self.engine.start()
        print("Car is ready to drive")


my_car = Car()
my_car.start_car()
````

### Understand the code

Look at this line:

```python
self.engine = Engine()
```

It creates an `Engine` object and stores it inside the `Car` object.

Now the `Car` can use the engine's method:

```python
self.engine.start()
```

Notice the difference:

- In inheritance, we use a parent class's methods directly through the child object.
- In composition, we access another object through an attribute, such as `self.engine`.

## 10. Composition with constructors and parameters

A class does not always need to create the object itself. We can pass an already-created object into its constructor.

This is useful because the class becomes flexible and can work with different objects.

````python
class Engine:
    def __init__(self, engine_type):
        self.engine_type = engine_type

    def start(self):
        print(self.engine_type, "engine started")


class Car:
    def __init__(self, brand, engine):
        self.brand = brand
        self.engine = engine  # Store the supplied Engine object

    def start_car(self):
        print(self.brand, "car:")
        self.engine.start()


petrol_engine = Engine("Petrol")
electric_engine = Engine("Electric")

car1 = Car("Toyota", petrol_engine)
car2 = Car("Tesla", electric_engine)

car1.start_car()
car2.start_car()
````

### Why is this useful?

The `Car` class does not need to know how every engine works internally. It only uses the engine object's `start()` method.

We can supply a different engine without rewriting the `Car` class.

## 11. Composition example: Computer and Processor

````python
class Processor:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def process(self):
        print(self.brand, "processor is processing at", self.speed, "GHz")


class Computer:
    def __init__(self, model, processor):
        self.model = model
        self.processor = processor

    def run_program(self):
        print("Computer:", self.model)
        self.processor.process()


cpu = Processor("Intel", 3.5)
computer = Computer("Dell XPS", cpu)

computer.run_program()
````

## 12. Composition example: Shopping Cart

A shopping cart has products. The cart does not need to inherit from `Product`.

Each product is a separate object, and the cart stores product objects in a list.

````python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price


class ShoppingCart:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def show_products(self):
        print("Products in cart:")
        for product in self.products:
            print(product.name, "-", product.price)

    def calculate_total(self):
        total = 0
        for product in self.products:
            total += product.price
        return total


laptop = Product("Laptop", 150000)
mouse = Product("Mouse", 2500)
keyboard = Product("Keyboard", 5000)

cart = ShoppingCart()
cart.add_product(laptop)
cart.add_product(mouse)
cart.add_product(keyboard)

cart.show_products()
print("Total:", cart.calculate_total())
````

### Why is this composition?

- A `ShoppingCart` has a list of products.
- A `Product` is not a type of `ShoppingCart`.
- The cart uses product objects to perform its work.

The relationship is **has-a**, not **is-a**.

## 13. Composition example: University and Department

````python
class Department:
    def __init__(self, name, head):
        self.name = name
        self.head = head

    def show_department(self):
        print("Department:", self.name)
        print("Head:", self.head)


class University:
    def __init__(self, name):
        self.name = name
        self.departments = []

    def add_department(self, department):
        self.departments.append(department)

    def show_university(self):
        print("University:", self.name)
        for department in self.departments:
            department.show_department()


cs = Department("Computer Science", "Dr. Ahmed")
business = Department("Business", "Dr. Sara")

university = University("Example University")
university.add_department(cs)
university.add_department(business)

university.show_university()
````

## 14. Composition vs aggregation (a useful distinction)

Both composition and aggregation are forms of object relationships, often discussed under the broader idea of *has-a* relationships.

- **Composition:** the containing object owns or controls the lifecycle of the contained object in the design.
- **Aggregation:** objects are associated, but the contained object can exist independently and may be shared.

Python does not enforce these lifecycle rules automatically. They are design decisions.

### Example: Composition-style design

A `House` creates and owns its `Room` objects.

````python
class Room:
    def __init__(self, name):
        self.name = name


class House:
    def __init__(self):
        self.rooms = [
            Room("Bedroom"),
            Room("Kitchen"),
            Room("Lounge")
        ]


house = House()

for room in house.rooms:
    print(room.name)
````

### Example: Aggregation-style design

A `Department` receives an existing `Teacher` object. The teacher can exist independently of the department.

````python
class Teacher:
    def __init__(self, name):
        self.name = name


class Department:
    def __init__(self, name, teacher):
        self.name = name
        self.teacher = teacher


teacher = Teacher("Dr. Ali")
department = Department("Computer Science", teacher)

print(department.name)
print(department.teacher.name)
print("Teacher still exists independently:", teacher.name)
````

## 15. Advantages and disadvantages of composition

### Advantages

- Classes are less tightly coupled to one another's inheritance structure.
- Components can be replaced or reused.
- Behavior can be delegated to specialized objects.
- The design can be easier to change as requirements grow.

### Disadvantages

- More objects and relationships may need to be created.
- You may write delegation methods to expose component behavior.
- A design with too many tiny objects can become unnecessarily complicated.

## 16. Inheritance vs composition

| Feature | Inheritance | Composition |
|---|---|---|
| Relationship | Is-a | Has-a |
| Main idea | Extend a class | Use another object |
| Reuse | Inherit methods and attributes | Delegate work to a component |
| Coupling | Often tighter | Often looser |
| Flexibility | Can be limited by hierarchy | Components can be exchanged |
| Example | Dog is an Animal | Car has an Engine |
| Python syntax | `class Dog(Animal)` | `self.engine = Engine()` |

### The most important question

Ask yourself:

- Is the new class genuinely a specialized type of the existing class? Consider inheritance.
- Does the new class contain or use another object to do its work? Consider composition.

## 17. Same problem solved using inheritance and composition

Suppose we are designing a notification system. We want different notification types, and each notification should send a message.

### Approach A — Inheritance

`EmailNotification` is a type of `Notification`.

````python
class Notification:
    def send(self, message):
        print("Sending notification:", message)


class EmailNotification(Notification):
    def send(self, message):
        print("Sending EMAIL:", message)


class SMSNotification(Notification):
    def send(self, message):
        print("Sending SMS:", message)


email = EmailNotification()
sms = SMSNotification()

email.send("Your order has shipped")
sms.send("Your order has shipped")
````

### Approach B — Composition

A notification service has a sender object. Different sender objects can be supplied.

````python
class EmailSender:
    def send(self, message):
        print("Sending EMAIL:", message)


class SMSSender:
    def send(self, message):
        print("Sending SMS:", message)


class NotificationService:
    def __init__(self, sender):
        self.sender = sender

    def notify(self, message):
        self.sender.send(message)


email_service = NotificationService(EmailSender())
sms_service = NotificationService(SMSSender())

email_service.notify("Your order has shipped")
sms_service.notify("Your order has shipped")
````

### Compare the two approaches

In the inheritance version, each notification class is a specialized notification type.

In the composition version, `NotificationService` delegates the sending operation to a sender object. The sender can be replaced when constructing the service.

Neither approach is automatically better. Choose based on the relationship and expected changes.

## 18. Polymorphism with inheritance and composition

**Polymorphism** means that different objects can be used through a common interface, while each object provides its own behavior.

### Polymorphism through inheritance

````python
class Shape:
    def area(self):
        raise NotImplementedError("Child classes must implement area()")


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2


shapes = [Rectangle(4, 5), Circle(3)]

for shape in shapes:
    print("Area:", shape.area())
````

### Polymorphism through composition

Python uses duck typing: an object can be used if it provides the method that the code expects. It does not necessarily need to inherit from a particular parent class.

````python
class PDFExporter:
    def export(self, data):
        print("Exporting data as PDF:", data)


class CSVExporter:
    def export(self, data):
        print("Exporting data as CSV:", data)


class Report:
    def __init__(self, exporter):
        self.exporter = exporter

    def generate(self, data):
        self.exporter.export(data)


pdf_report = Report(PDFExporter())
csv_report = Report(CSVExporter())

pdf_report.generate("Monthly sales")
csv_report.generate("Monthly sales")
````

## 19. A practical decision guide

Use inheritance when:

1. There is a clear *is-a* relationship.
2. Child classes share meaningful common behavior.
3. You want a common parent interface and specialized implementations.

Use composition when:

1. There is a clear *has-a* relationship.
2. One class needs another object to perform a task.
3. You want to change or replace a component independently.
4. You want to avoid a deep or unnatural inheritance hierarchy.

**Remember:** Composition is often a good default for assembling behavior, but inheritance remains useful for genuine type hierarchies.

## 27. Solution: Employee Management

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show_details(self):
        print("Name:", self.name)
        print("Salary:", self.salary)


class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def manage_team(self):
        print(self.name, "is managing the", self.department, "team")


manager = Manager("Ali", 150000, "Engineering")
manager.show_details()
print("Department:", manager.department)
manager.manage_team()
````

## 28. Solution: Vehicles

````python
class Vehicle:
    def start(self):
        print("Vehicle is starting")


class Car(Vehicle):
    def start(self):
        print("Car engine started")


class Motorcycle(Vehicle):
    def start(self):
        print("Motorcycle engine started")


car = Car()
motorcycle = Motorcycle()

car.start()
motorcycle.start()
````

## 29. Solution: Library and Book

````python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def show_books(self):
        for book in self.books:
            print(book.title, "by", book.author)


book1 = Book("Python Basics", "A. Writer")
book2 = Book("OOP Concepts", "B. Writer")
book3 = Book("Clean Code", "C. Writer")

library = Library()
library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

library.show_books()
````

## 30. Solution: Mobile Phone and Battery

````python
class Battery:
    def __init__(self, capacity):
        self.capacity = capacity

    def charge(self):
        print("Battery capacity:", self.capacity, "mAh")


class MobilePhone:
    def __init__(self, brand, battery):
        self.brand = brand
        self.battery = battery

    def use_phone(self):
        print("Using", self.brand, "phone")
        self.battery.charge()


phone_battery = Battery(5000)
phone = MobilePhone("Samsung", phone_battery)

phone.use_phone()
````
