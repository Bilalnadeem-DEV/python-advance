```python
# Multiple Aggregators
```

## Suppose you want:

Average flavor rating.

Maximum flavor rating.

Minimum flavor rating.

Average texture rating.

Total rating sum.



```python
df.groupby('Base Flavor').agg({
    'Flavor Rating': ['mean', 'max', 'min'],
    'Texture Rating': ['mean', 'max', 'min'],
    'Total Rating': ['mean', 'sum']
})
```

## Named aggregation

```python
df.groupby('Base Flavor').agg(
    Average_Flavor=('Flavor Rating', 'mean'),
    Maximum_Flavor=('Flavor Rating', 'max'),
    Minimum_Flavor=('Flavor Rating', 'min'),
    Average_Texture=('Texture Rating', 'mean'),
    Average_Total=('Total Rating', 'mean'),
    Total_Score=('Total Rating', 'sum')
)
```
