## 1. What is Pandas?

**Pandas** is a Python library for working with structured and tabular data. It is commonly used for:

### The two main data structures

| Structure | Dimensions | Description |
|---|---:|---|
| `Series` | 1D | A labeled sequence of values |
| `DataFrame` | 2D | A table made of rows and columns |

A DataFrame can be thought of as a collection of Series that share the same row index.

## 10. Handling missing values

Missing values are commonly represented by `NaN`, `None`, or `pd.NA`. They can affect calculations and should be investigated before analysis.

````python
messy = pd.DataFrame({
    "Name": ["Ali", "Sara", "Hamza", "Ayesha", "Bilal", None],
    "Age": [20, np.nan, 19, 22, np.nan, 21],
    "Marks": [85, 92, np.nan, 95, 88, 70],
    "City": ["Lahore", "Karachi", None, "Lahore", "Multan", "Karachi"]
})
messy
````

````python
messy.isna()                 # True where missing
messy.isna().sum()           # missing count per column
messy.notna().sum()          # non-missing count per column
messy.isna().sum().sum()     # total missing cells
````

````python
# Remove rows/columns with missing values
messy.dropna()                       # rows with any missing value
messy.dropna(subset=["Marks"])       # rows missing Marks
messy.dropna(axis=1)                 # columns containing missing values
messy.dropna(how="all")              # rows where every value is missing
````

````python
# Fill missing values
filled = messy.copy()
filled["Age"] = filled["Age"].fillna(filled["Age"].median())
filled["Marks"] = filled["Marks"].fillna(filled["Marks"].mean())
filled["City"] = filled["City"].fillna("Unknown")
filled["Name"] = filled["Name"].fillna("Not Provided")
filled
````

````python
# Forward fill and backward fill
messy.ffill()
messy.bfill()
````

**Choosing a strategy**
- Numeric data: mean or median may be appropriate, depending on outliers and distribution.
- Categorical data: mode or a meaningful label such as `Unknown`.
- Time-series data: forward/backward filling may make sense when the order is meaningful.
- Do not automatically fill every missing value; first ask why it is missing.

## 18. GroupBy — split, apply, combine

`groupby()` divides data into groups, performs calculations for each group, and combines the results.

````python
sales = pd.DataFrame({
    "Region": ["North", "South", "North", "East", "South", "East", "North", "South"],
    "Product": ["Laptop", "Phone", "Phone", "Laptop", "Laptop", "Phone", "Tablet", "Tablet"],
    "Units": [5, 8, 6, 3, 4, 9, 7, 2],
    "Revenue": [250000, 160000, 120000, 150000, 200000, 180000, 140000, 40000]
})
sales
````

````python
sales.groupby("Region")["Revenue"].sum()
````

````python
sales.groupby(["Region", "Product"])[["Units", "Revenue"]].sum()
````

````python
sales.groupby("Region").agg(
    total_revenue=("Revenue", "sum"),
    average_revenue=("Revenue", "mean"),
    total_units=("Units", "sum"),
    transaction_count=("Product", "count")
)
````

````python
# Transform returns a result aligned to the original rows
sales["RegionAverage"] = sales.groupby("Region")["Revenue"].transform("mean")
sales
````

````python
# Filter entire groups based on a group-level condition
sales.groupby("Region").filter(lambda group: group["Revenue"].sum() > 250000)
````

## 20. Merging and joining DataFrames

````python
students_info = pd.DataFrame({
    "StudentID": [1, 2, 3, 4],
    "Name": ["Ali", "Sara", "Hamza", "Ayesha"],
    "DepartmentID": [10, 20, 10, 30]
})
departments = pd.DataFrame({
    "DepartmentID": [10, 20, 30],
    "DepartmentName": ["CS", "SE", "AI"]
})
````

````python
# Inner join: only matching keys
pd.merge(students_info, departments, on="DepartmentID", how="inner")
````

````python
# Left join: keep all rows from the left DataFrame
pd.merge(students_info, departments, on="DepartmentID", how="left")
````

````python
# Right and outer joins
pd.merge(students_info, departments, on="DepartmentID", how="right")
pd.merge(students_info, departments, on="DepartmentID", how="outer")
````

````python
# Different key names
pd.merge(
    students_info,
    departments,
    left_on="DepartmentID",
    right_on="DepartmentID",
    validate="many_to_one"
)
````

````python
# Merge when both DataFrames have overlapping non-key column names
one = pd.DataFrame({"ID": [1, 2], "Score": [80, 90]})
two = pd.DataFrame({"ID": [1, 2], "Score": [85, 95]})
pd.merge(one, two, on="ID", suffixes=("_exam1", "_exam2"))
````

**Join types**
- `inner`: matching keys only.
- `left`: all left rows, matching right rows where available.
- `right`: all right rows.
- `outer`: all keys from both sides.

Use `validate` when you expect a relationship such as one-to-one or many-to-one; it can catch accidental row multiplication.

## 24. Working with outliers

````python
outlier_df = pd.DataFrame({"Salary": [30000, 32000, 35000, 37000, 40000, 42000, 500000]})
q1 = outlier_df["Salary"].quantile(0.25)
q3 = outlier_df["Salary"].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outlier_df["IsOutlier"] = (
    (outlier_df["Salary"] < lower_bound) |
    (outlier_df["Salary"] > upper_bound)
)
outlier_df
````

The IQR rule is a screening technique, not proof that a value is wrong. An unusually high salary may be valid. Investigate context before removing or changing outliers.

## 27. Window functions: rolling, expanding, and cumulative

````python
daily = pd.DataFrame({
    "Day": pd.date_range("2025-01-01", periods=7),
    "Sales": [10, 15, 12, 20, 18, 25, 22]
}).set_index("Day")
daily
````

````python
daily["Rolling3DayMean"] = daily["Sales"].rolling(window=3).mean()
daily["CumulativeSales"] = daily["Sales"].cumsum()
daily["ExpandingMean"] = daily["Sales"].expanding().mean()
daily
````

- `rolling(3).mean()`: calculation over the latest three observations.
- `expanding().mean()`: calculation from the first observation through the current one.
- `cumsum()`: cumulative sum.
- Other cumulative methods include `cummin()`, `cummax()`, and `cumprod()`.

## 30. Efficient Pandas habits

1. **Prefer vectorization:** `df["total"] = df["price"] * df["quantity"]` is usually better than looping row by row.
2. **Read only needed columns:** use `usecols` with `read_csv`.
3. **Set data types intentionally:** avoid storing numbers as strings.
4. **Use categorical types** for repeated labels when suitable.
5. **Avoid chained assignment:** use `.loc[row_condition, "column"] = value`.
6. **Inspect memory:** `df.memory_usage(deep=True)` and `df.info(memory_usage="deep")`.
7. **Use `copy()`** when making an independent working DataFrame.
8. **Validate joins** with `validate=` where possible.
9. **Use `groupby().agg()`** for clear summaries.
10. **Do not assume missing values or outliers are errors**; investigate them.

## 31. Common errors and how to fix them

| Error / issue | Typical reason | How to investigate |
|---|---|---|
| `KeyError: 'Marks'` | Column name does not exist or has extra spaces | `print(df.columns)` |
| `AttributeError` | Calling a method on the wrong object/type | `type(value)`, `df.dtypes` |

| Unexpected `NaN` after merge | Keys do not match exactly | Inspect key types, spaces, and spelling |

# 32. Guided mini-project: Student Performance Analysis

You are given student records. Your job is to clean the data, calculate performance metrics, and answer questions for the instructor.

````python
student_data = pd.DataFrame({
    "StudentID": [101, 102, 103, 104, 105, 106, 107, 108, 108],
    "Name": ["Ali", "Sara", "Hamza", "Ayesha", "Bilal", "Noor", "Usman", "Hira", "Hira"],
    "Department": ["CS", "SE", "CS", "AI", "SE", "AI", "CS", "AI", "AI"],
    "Quiz": [8, 9, 6, 10, 7, np.nan, 5, 9, 9],
    "Midterm": [75, 88, 62, 95, 70, 82, np.nan, 91, 91],
    "Final": [80, 92, 65, 98, 74, 85, 60, 94, 94],
    "Attendance": [90, 95, 70, 98, 75, 88, 60, 96, 96]
})
student_data
````

### Project tasks (complete independently)
1. Inspect the dataset: shape, columns, data types, and summary statistics.
2. Identify duplicate records. Remove exact duplicate rows.
3. Check missing values in every column.
4. Fill missing Quiz and Midterm marks using a justified strategy.
5. Create `Total` as Quiz (out of 10), Midterm (out of 100), and Final (out of 100), converted to a percentage out of 210.
6. Create `Percentage` and `Grade` columns. Define your own grade boundaries and document them.
7. Find the top five students by percentage.
8. Calculate average percentage by department.
9. Find the department with the highest average percentage.
10. Find students with attendance below 75%.
11. Count students in each grade.
12. Create a bar chart of average percentage by department.
13. Export the cleaned dataset to `student_performance_cleaned.csv`.

**Challenge:** Create a second summary table with department, student count, average percentage, highest percentage, and lowest percentage.

### Project solution — reveal only after attempting the tasks

````python
# 1. Copy and inspect
clean = student_data.copy()
print(clean.shape)
clean.info()
display(clean.describe(include="all"))

# 2. Remove exact duplicate rows
clean = clean.drop_duplicates()

# 3. Missing-value report
print(clean.isna().sum())

# 4. Fill missing marks with the relevant column median
for col in ["Quiz", "Midterm"]:
    clean[col] = clean[col].fillna(clean[col].median())

# 5. Total out of 210 and percentage
clean["Total"] = clean["Quiz"] + clean["Midterm"] + clean["Final"]
clean["Percentage"] = clean["Total"] / 210 * 100

# 6. Grade boundaries (example policy)
def assign_grade(p):
    if p >= 90:
        return "A"
    elif p >= 80:
        return "B"
    elif p >= 70:
        return "C"
    elif p >= 60:
        return "D"
    return "F"

clean["Grade"] = clean["Percentage"].apply(assign_grade)

# 7. Top five
display(clean.nlargest(5, "Percentage"))

# 8–9. Department summary
department_summary = clean.groupby("Department").agg(
    StudentCount=("StudentID", "count"),
    AveragePercentage=("Percentage", "mean"),
    HighestPercentage=("Percentage", "max"),
    LowestPercentage=("Percentage", "min")
).sort_values("AveragePercentage", ascending=False)

display(department_summary)

# 10. Low attendance
display(clean[clean["Attendance"] < 75])

# 11. Grade counts
display(clean["Grade"].value_counts())

# 12. Plot
department_summary["AveragePercentage"].plot(
    kind="bar", title="Average Percentage by Department"
)
plt.ylabel("Average percentage")
plt.show()

# 13. Export
clean.to_csv("student_performance_cleaned.csv", index=False)
````
