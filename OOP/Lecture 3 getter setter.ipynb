```python
class Car:
    make = ""
    Color = ""
    horn = False
    
    #Create/ initialize
    def __init__(self, make, color):
        self.make = make
        self.color = color
        
    #Read -> Getters
    def getMake(self):
        return self.make
    
    # Update -> Setter
    def setMake(self, value):
        self.make = value
    
    # Delete -> Delete
    def deleteMake(self):
        self.make = ""
        
    #Read -> Getters
    def getHorn(self):
        if self.horn == False:            
            print("Horn is off")
        else:
            print("Horn is ON")
        return self.horn
    
    # Update -> Setter
    def setHorn(self, value):
        self.horn = value
    
    # Delete -> Delete
    def resetHorn(self):
        self.horn = False
        
```
