```python
class Bank:
    __balance = 0
    def __init__(self, initial_balance=0):
        # Private variable to store the account balance
        self.__balance = initial_balance

    def deposit(self, amount):
        if amount > 0:
            self.__update_balance(amount)
            print(f"Deposited: ${amount:.2f}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > 0:
            if self.__check_sufficient_balance(amount):
                self.__update_balance(-amount)
                print(f"Withdrew: ${amount:.2f}")
            else:
                print("Insufficient funds.")
        else:
            print("Withdrawal amount must be positive.")

    def get_balance(self):
        return self.__balance
    
    # Private method to update balance
    def __update_balance(self, amount):
        self.__balance += amount

    # Private method to check if sufficient balance exists
    def __check_sufficient_balance(self, amount):
        return self.__balance >= amount
```

```python
class Bank:
    def __init__(self, initial_balance=0):
        # Protected variable to store the account balance
        self._balance = initial_balance

    def deposit(self, amount):
        if amount > 0:
            self._update_balance(amount)
            print(f"Deposited: ${amount:.2f}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > 0:
            if self._check_sufficient_balance(amount):
                self._update_balance(-amount)
                print(f"Withdrew: ${amount:.2f}")
            else:
                print("Insufficient funds.")
        else:
            print("Withdrawal amount must be positive.")

    def get_balance(self):
        return self._balance

    # Protected method to update balance
    def _update_balance(self, amount):
        self._balance += amount

    # Protected method to check if sufficient balance exists
    def _check_sufficient_balance(self, amount):
        return self._balance >= amount


# Example subclass to demonstrate protected access
class SavingsAccount(Bank):
    def add_interest(self, rate):
        interest = self._balance * rate
        self._update_balance(interest)
        print(f"Interest added: ${interest:.2f}")


account = Bank(100)  # Initialize the account with $100
account.deposit(50)   # Deposit $50
print(f"Current Balance: ${account.get_balance():.2f}")  # Get balance
account.withdraw(30)  # Withdraw $30
print(f"Current Balance: ${account.get_balance():.2f}")  # Get balance

# Using the SavingsAccount subclass
savings_account = SavingsAccount(200)
savings_account.add_interest(0.05)  # Add 5% interest
print(f"Current Balance in Savings Account: ${savings_account.get_balance():.2f}")
```
