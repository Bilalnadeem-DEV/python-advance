Step 2: Solve each module

```python
# Declare all variables and initialize them
# Good programming practice
var1 = 0
var2 = 0

# Modules/ methods/ functions

def takeInputs():
    global var1
    global var2
    var1 = input("Enter first number:  ")
    var2 = input("Enter second number: ")
    typeCasting()

def typeCasting():
    global var1
    global var2
    try:
        var1 = int(var1)
        var2 = int(var2)
    except:
        print("Please provide the numeric input")
        takeInputs()
        
def addition(v1, v2):
    result = v1 + v2
    print("Sum pf numbers is: ", result)
            
def subtraction(v1, v2):
    result = v1 - v2
    print("Subtraction pf numbers is: ", result)

def multiplication(v1, v2):
    result = v1 * v2
    print("Multiplication pf numbers is: ", result)
    
def division(v1, v2):
    result = v1 / v2
    print("Division pf numbers is: ", result)
```

```python
while True:   
    takeInputs()
    addition(var1, var2)
    subtraction(var1, var2)
    multiplication(var1, var2)
    division(var1, var2)
    print("------------------------------------------------")
    print("------------------------------------------------")
```
