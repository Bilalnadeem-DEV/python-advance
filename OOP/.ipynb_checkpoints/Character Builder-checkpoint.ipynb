```python
class BaseCharacter:
    def __init__(self, name, age, gender, HP):
        self.name = name
        self.age = age
        self.gender = gender
        self.HP = HP

    def showDetails(self):
        print("Character Name: .. ", self.name)
        print("Character Age: .. ", self.age)
        print("Character Gender: .. ", self.gender)
        print("Character HP: .. ", self.HP)
        

class WeaponExpert(BaseCharacter):
    def __init__(self, primaryWeapon, primaryWeaponAmmo, secondaryWeapon, secondaryWeaponAmmo, meleeWeapon, name, age, gender, HP):
        BaseCharacter.__init__(self, name, age, gender, HP)
        self.primaryWeapon = primaryWeapon
        self.secondaryWeapon = secondaryWeapon
        self.primaryWeaponAmmo = primaryWeaponAmmo
        self.meleeWeapon = meleeWeapon
        self.secondaryWeaponAmmo = secondaryWeaponAmmo
        self.meleeWeapon = meleeWeapon
        
    
    def showDetails(self):
        BaseCharacter.showDetails()
        print("Primary Weapon:.. ", self.primaryWeapon)
        print("Secondary Weapon: .. ", self.secondaryWeapon)
        print("Melee weapon: .. ", self.meleeWeapon)
    
    
    # Polymprphism
    def attack(self, target, weaponType = "Primary", damage=40):
        
        if weaponType == "Primary": 
            if self.primaryWeaponAmmo >=1:
                self.primaryWeaponAmmo -=self.primaryWeaponAmmo
                print(f"Attacked with {self.primaryWeapon} weapon")
                print(f"Damage taken by {self.primaryWeapon} weapon is {damage}")
                
            else:
                print("Primary Weapon on low ammo")
        elif weaponType == "Secondary":
             if self.secondaryWeaponAmmo >=1:
                self.secondaryWeaponAmmo -=self.secondaryWeaponAmmo
                print(f"Attacked with {self.secondaryWeapon} weapon")
                print(f"Damage taken by {self.secondaryWeapon} weapon is {damage}")
            else:
                print("Secondary Weapon on low ammo")
        else: 
            print(f"Attacked with {self.meleeWeapon} weapon")
            print(f"Damage taken by {self.meleeWeapon} weapon is {damage}")
        
    def reloadAllWeapons(self):
        self.primaryWeaponAmmo = 100
        self.secondaryWeaponAmmo = 100
        
        

class HeavyArmour(BaseCharacter):
    def __init__(self, armour, name, age, gender, HP):
        BaseCharacter.__init__(self, name, age, gender, HP)
        self.armour = armour
    
    def showDetails(self):
        BaseCharacter.showDetails()
        print("Armour:.. ", self.armour)
    
    def addArmour(self, armourValue):
        if self.armour == 100:
            return
        elif self.armour + armour > 100:
            self.armour = 100
        elif:
            self.armour = self.armour + armourValue

            
class TacticalMind(BaseCharacter):
    def __init__(self, tacticalMind, name, age, gender, HP):
        BaseCharacter.__init__(self, name, age, gender, HP)
        self.tacticalMind = tacticalMind
    
    def showDetails(self):
        BaseCharacter.showDetails()
        print("Tactical Mind:.. ", self.tacticalMind)
    
    def plan_strategy_1(self):
        print(f"{self.name} finds cover and studies the enemy.")
    
    def plan_strategy_1(self):
        print(f"{self.name} look for the health")
        


class Gunner(WeaponExpert, HeavyArmour, TacticalMind):
    def __init__(self, primaryWeapon, primaryWeaponAmmo, secondaryWeapon, secondaryWeaponAmmo, meleeWeapon, armour, tacticalMind,
                 name, age, gender, HP):
        BaseCharacter.__init__(self, name, age, gender, HP)

        # WeaponExpert attributes
        self.primaryWeapon = primaryWeapon
        self.primaryWeaponAmmo = primaryWeaponAmmo
        self.secondaryWeapon = secondaryWeapon
        self.secondaryWeaponAmmo = secondaryWeaponAmmo
        self.meleeWeapon = meleeWeapon
        self.armour = armour
        self.tacticalMind = tacticalMind
        
        def showDetails(self):
            BaseCharacter.showDetails()
            print("Primary Weapon:.. ", self.primaryWeapon)
            print("Secondary Weapon: .. ", self.secondaryWeapon)
            print("Melee weapon: .. ", self.meleeWeapon)
            print("Armour:.. ", self.armour)
            print("Tactical Mind:.. ", self.tacticalMind)
            


# Create your character
sniper = Sniper("Raven", 23, "M", 100)

# sniper.attack("Zombie")
# sniper.attack("Zombie", "Sniper Rifle", 90)
# sniper.plan_strategy()

# sniper.health = 60
# sniper.heal()

# # Polymorphism: same method call, different behavior
# characters = [
#     BaseCharacter("Alex"),
#     WeaponExpert("Sam"),
#     sniper
# ]

# for character in characters:
#     character.use_ability()
```
