## Student example with constructor

```python
class Student:
    id = 0
    name = ""
    phoneNumber = 0
    email = ""
    
    def __init__(self, id, name, phoneNumber, email):
        self.id = id
        self.name = name
        self.phoneNumber = phoneNumber
        self.email = email
```

```python
s1 = Student(10, "Thor", "111111", "thor@email.com")
s2 = Student(11, "Hulk", "444444", "hulk@email.com")
s3 = Student(12, "Cap", "222222", "cap@email.com")
s4 = Student(13, "Tonny", "55555", "tonny@email.com")
```

```python
s1.name
```

```python
s2.name
```

## Car example 

```python
class Car:
    name = ""
    year = "" 
    make = ""
    headLights = False
    currentSpeed = 0
    isLocked = False
    
    def __init__(self, name, year, make):
        self.name = name
        self.year = year
        self.make = make
    
    def turnOnHeadLights(self):
        self.headLights = True
    
    def turnOffHeadLights(self):
        self.headLights = False
    
    def accelerate(self):
        self.currentSpeed = self.currentSpeed + 10
    
    def decelerate(self):
        if currentSpeed > 10:
            self.currentSpeed = self.currentSpeed - 10
        else:
            self.currentSpeed = 0
    
    # Getter and setter of class
    def getName(self):
        print(self.name)
    def setName(self, name):
        self.name = name
```

```python
tucson = Car("Tucson", "2024", "Hyundai")
sportage = Car("Sportage", "2024", "Kia")
cross = Car("Cross", "2024", "Toyota")
```

```python
class Student:
#     Attributes
    name = ""
    email = ""
    
    # Constructor
    def __init__(self, name, email):
        self.name = name
        self.email = email
    
    def dropOut(self):
        print("Student dropped out, Thanks for trying")
        
    def editName(self):
        newName = input("Enter name")
        self.name = newName
        
```

```python
# Weapon example

class Weapon:
    isFiring = False
    isAlive = True
    
    def shoot(self):
        self.isFiring = True
    
    def declareDead(self):
        self.isAlive = False
        
```

```python
w1 = Weapon()
w2 = Weapon()
w3 = Weapon()
w4 = Weapon()
```

```python
print(w1.isFiring)
print(w2.isFiring)
print(w3.isFiring)
print(w4.isFiring)
```

```python
w2.shoot()
w1.declareDead()
```

```python
w1.isAlive
```

```python
class Car:
    def __init__(self, name):
        self.name = name
        self.horn = False
        
    def getName(self):
        return self.name
    def getHorn(self):
        return self.horn
    
    def setName(self, name):
        self.name = name
    def setHorn(self,horn):
        if horn == True:
            print("Horn Turned on for 2 seconds")
            self.horn = horn
            self.setHorn(False)
        else:
            self.horn = False
    
```

```python
def takeInput():
    name = input("Enter your name ")
    age = input("Enter your age   ")
    typeCase(name, age)
#     createClass(name, intAge)
    
def typeCase(name, age):
    try:
        intTyprAge = int(age)
        createClass(intTyprAge, name)
    except ValueError:
        takeInput()
```

```python
class Students:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def show_details(self):
        print("=========================================")
        print(self.name)
        print(self.age)
        print("=========================================")
```

```python
def createClass(name, age):
    obj = Students(name, age)
    students.append(obj)
```

```python
for i in range(5):
    takeInput()
```

```python
for i in range(len(students)):
    students[i].show_details()
```
