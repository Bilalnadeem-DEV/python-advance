# Encapsulation and polymorphism usecase

```python
def printInfo(a, *args, b = 0, c = None ):
    
    print("A = ", a)
    
    if args:
        for data in args:
            print("ARGS", data)
    
    if b != 0:
        print("B ", b)
        
    if c != None:
        print("C ", c)
        
printInfo(10, 20,30,40,50, b = 60, c = 70)
```

```python
def fibonacci(n):
    fib_series = []
    a = 0
    b = 1
    for _ in range(n):
        fib_series.append(a)
        
        # Update a and b without using one-line swap
        temp = a
        a = b
        b = temp + b
        
    return fib_series

# Example: Print the first 10 terms of the Fibonacci series
print(fibonacci(10))
```

```python
class Account:
    def __init__(self, account_number, owner, balance=0):
        self.__account_number = account_number  # Private attribute
        self.__owner = owner                    # Private attribute
        self.__balance = balance                # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"{amount} deposited. New balance: {self.__balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"{amount} withdrawn. New balance: {self.__balance}")
        else:
            print("Insufficient funds.")

    def get_balance(self):
        return self.__balance

    def account_details(self):
        return f"Account Number: {self.__account_number}, Owner: {self.__owner}, Balance: {self.__balance}"


class SavingsAccount(Account):
    def __init__(self, account_number, owner, balance=0, interest_rate=0.02):
        super().__init__(account_number, owner, balance)
        self.__interest_rate = interest_rate  # Private attribute

    def apply_interest(self):
        interest = self.get_balance() * self.__interest_rate
        self.deposit(interest)
        print(f"Interest of {interest} applied.")


class CheckingAccount(Account):
    def __init__(self, account_number, owner, balance=0, overdraft_limit=0):
        super().__init__(account_number, owner, balance)
        self.__overdraft_limit = overdraft_limit  # Private attribute

    def withdraw(self, amount):
        if amount <= self.get_balance() + self.__overdraft_limit:
            super().l(amount)  # Call the parent withdraw method
        else:
            print("Withdrawal amount exceeds overdraft limit.")


# Example usage
def main():
    # Create accounts
    savings_account = SavingsAccount("SA123", "Alice", 1000)
    checking_account = CheckingAccount("CA456", "Bob", 500, overdraft_limit=200)

    # Perform operations on Savings Account
    print(savings_account.account_details())
    savings_account.deposit(200)
    savings_account.apply_interest()
    print(savings_account.account_details())

    print("\n")

    # Perform operations on Checking Account
    print(checking_account.account_details())
    checking_account.withdraw(600)  # Should succeed due to overdraft
    checking_account.withdraw(200);   # Should succeed
    checking_account.withdraw(500);   # Should fail
    print(checking_account.account_details())


if __name__ == "__main__":
    main()
```

# Strategy Pattern

```python
class PaymentStrategy:
    def pay(self, amount):
        raise NotImplementedError("This method should be overridden.")

class CreditCardPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paying {amount} using Credit Card.")

class PayPalPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paying {amount} using PayPal.")

class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def checkout(self, payment_strategy: PaymentStrategy):
        total = sum(self.items)
        payment_strategy.pay(total)

# Usage
cart = ShoppingCart()
cart.add_item(100)
cart.add_item(200)

# Pay using different strategies
cart.checkout(CreditCardPayment())  # Paying 300 using Credit Card.
cart.checkout(PayPalPayment())      # Paying 300 using PayPal.
```

Base Class (PaymentStrategy): Defines an interface for different payment methods.
Concrete Classes: CreditCardPayment and PayPalPayment implement the pay() method differently.
Polymorphism in Action: The checkout method in ShoppingCart accepts any PaymentStrategy, allowing different payment methods to be used interchangeably.

# Factory Pattern

```python
class Shape:
    def draw(self):
        raise NotImplementedError("Subclasses must implement this method.")

class Circle(Shape):
    def draw(self):
        return "Drawing a Circle."

class Square(Shape):
    def draw(self):
        return "Drawing a Square."

class ShapeFactory:
#     @staticmethod
    def get_shape(shape_type):
        if shape_type == "CIRCLE":
            return Circle()
        elif shape_type == "SQUARE":
            return Square()
        else:
            raise ValueError("Unknown shape type.")

# Usage
shapes = ["CIRCLE", "SQUARE"]
for shape_type in shapes:
    shape = ShapeFactory.get_shape(shape_type)
    print(shape.draw())
```

Base Class (Shape): Defines an interface with the draw() method.
Concrete Classes: Circle and Square implement the draw() method.
Factory Method: The ShapeFactory creates instances of Shape without exposing the instantiation logic to the client.

# Observer Pattern

```python
class Observer:
    def update(self, message):
        raise NotImplementedError("Subclasses must implement this method.")

class EmailObserver(Observer):
    def update(self, message):
        print(f"Email: {message}")

class SMSObserver(Observer):
    def update(self, message):
        print(f"SMS: {message}")

class Subject:
    def __init__(self):
        self.observers = []

    def attach(self, observer: Observer):
        self.observers.append(observer)

    def notify(self, message):
        for observer in self.observers:
            observer.update(message)

# Usage
subject = Subject()
subject.attach(EmailObserver())
subject.attach(SMSObserver())

subject.notify("New event happened!")  
# Output:
# Email: New event happened!
# SMS: New event happened!
```

Base Class (Observer): Defines the update() method that all observers must implement.
Concrete Observers: EmailObserver and SMSObserver implement the update() method differently.
Subject Class: Manages a list of observers and notifies them of events, demonstrating polymorphism.
