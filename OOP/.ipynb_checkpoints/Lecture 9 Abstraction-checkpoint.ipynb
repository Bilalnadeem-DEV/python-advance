```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    
    engineOn = "OFF"
    
    @abstractmethod
    def turnEngineOn(self, engineOn):
        pass
    
    @abstractmethod
    def turnEngineOFF(self, engineOn):
        pass     
    
class Car(Vehicle):
    
    def turnEngineOn(self, engineOn):
        self.engineOn = engineOn
        print("Engine turned on")
        
    def turnEngineOFF(self, engineOn):
        self.engineOn = engineOn
        print("Engine turned off")
    
    def random(self):
        print("This is random function")
    
```

```python
c1 = Car()
```

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
        
    @abstractmethod
    def startEngine(self):
        pass
    @abstractmethod
    def setGear(self):
        pass
    @abstractmethod
    def breaking(self):
        pass
    
class Car(Vehicle):
    make = ""
    
    def startEngine(self):
        print("Car engine started")
    def setGear(self):
        print("Set gear called")
    def breaking(self):
        print("Breaking called")
    
```

```python
from abc import ABC, abstractmethod

class Car(ABC):
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    @abstractmethod
    def start_engine(self):
        pass

    @abstractmethod
    def stop_engine(self):
        pass

    @abstractmethod
    def drive(self):
        pass

class ElectricCar(Car):
    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self._battery_level = 100
        self._engine_started = False

#     def start_engine(self):
#         if not self._engine_started:
#             self._engine_started = True
#             print(f"The electric engine of {self.make} {self.model} is now started.")
#         else:
#             print("The electric engine is already running.")

    def stop_engine(self):
        if self._engine_started:
            self._engine_started = False
            print(f"The electric engine of {self.make} {self.model} has been stopped.")
        else:
            print("The electric engine is already off.")

    def drive(self):
        if self._engine_started:
            print(f"Driving the electric {self.make} {self.model}. Battery level: {self._battery_level}%.")
        else:
            print("You need to start the engine first.")

# Using the ElectricCar class
my_electric_car = ElectricCar("Tesla", "Model S", 2023)
my_electric_car.start_engine()  # Starts the engine
my_electric_car.drive()         # Drives the car
my_electric_car.stop_engine()   # Stops the engine
```
