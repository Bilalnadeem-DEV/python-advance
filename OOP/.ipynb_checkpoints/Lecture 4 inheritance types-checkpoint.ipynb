```python
class Vehicle:
    make = ""
    model = ""
    year = ""
    
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        
class Car:
    engineStatus = "OF"
    lockStatus = "Locked"
    
    def __init__(self, engineStatus, lockStatus):
        self.engineStatus = engineStatus
        self.lockStatus = lockStatus
        
class ElectricCar: 
    batteryPercentage = "100%"
    
    def __init__(self, batteryPercentage):
        self.batteryPercentage = batteryPercentage

class LucuryVehicle:
    ADASenabled = "Enabled"
    
    def __init__(self, ADASenabled):
        self.ADASenabled = ADASenabled
        
class Sedan(Vehicle, Car):
    
    def __init__(self, make, model, year, engineStatus, lockStatus):
        Vehicle.__init__(self, make, model, year)
        Car.__init__(self, engineStatus, lockStatus)
        
        
class LuxuryElectric(Vehicle, Car, ElectricCar, LucuryVehicle):
    autoPilot = "ON"
    
    def __init__(self, make, model, year, engineStatus, lockStatus, batteryPercentage, ADASenabled, autoPilot):
        Vehicle.__init__(self, make, model, year)
        Car.__init__(self, engineStatus, lockStatus)
        ElectricCar.__init__(self, batteryPercentage)
        LucuryVehicle.__init__(self, ADASenabled)
                
        self.autoPilot = autoPilot
        
```

```python
civic = Sedan("Civic", "X", "2019", "OFF", "Locked")
```

```python
cyberTruck = LuxuryElectric("Tesla", "Cyber Truck", "2023", "ON", "Unlocked", "60%", "ADAS ON", "ON")
```

```python
cyberTruck.
```

```python
class Vehicle:
    def __init__(self, make, model):
        self.Make = make
        self.model = model
    
class Car(Vehicle):
    def __init__(self, sunRoof, AC):
        self.sunRoof = sunRoof
        self.AC = AC
        
    def getSunRoof(self):
        return sunRoof
    
    def getSunRoof(self):
        return sunRoof

class Suv(Car):
    def __init__(self, fbf, sunRoof, AC, make, model):
        Car.__init__(self, sunRoof, AC)
        Vehicle.__init__(self, make, model)
        self.fbf = fbf
        
        
class Phev(Car):
    def __init__(self, battery, sunRoof, AC, make, model):
        Car.__init__(self, sunRoof, AC)
        Vehicle.__init__(self, make, model)
        self.battery = battery

    def getBattery(self):
        return battery
    
    
```

```python
class Vehicle:
    make = ""
    model = ""
    horn = False
    
    def __init__(self, make, model):
        self.make = make
        self.model = model
    
    def getMake(self):
        return self.make
    def getModel(self):
        return self.model
    def getHorn(self):
        return self.horn
    
    def setMake(self, make):
        self.make = make
    def setModel(self, model):
        self.model = model
    def setHorn(self, horn):
        self.horn = horn


class Car(Vehicle):
    roof = True
    trunk = ""
    def __init__(self, roof, trunk, make, model):
        Vehicle.__init__(self,make, model)
        self.roof = roof
        self.trunk = trunk
    
    def getRoof(self):
        return self.roof
    def getTrunk(self):
        return self.trunk
    
    def setRoof(self, roof):
        self.roof = roof
    def setTrunk(self, trunk):
        self.trunk = trunk
        
class EV(Car):
    battery = ""
    def __init__(self, battery, roof, trunk, make, model):
        Car.__init__(self, roof, trunk, make, model)
        self.battery = battery

class SuvEV(EV):
    fbf = ""
    def __init__(self, fbf, battery, roof, trunk, make, model):
        EV.__init__(self, battery, roof, trunk, make, model)
        self.fbf = fbf
    
```

```python
civic = Car("Silver Roof","400L", "Honda", "Civic Oriel 2023" )
```

```python
ev = EV("100KW", "true", "500L", "MG", "HS")
```

```python
class Vehicle:
    def __init__(self, name, make, year):
        self.name = name
        self.make = make
        self.year = year
        
class Car(Vehicle):
    def __init__(self, sunRoof, staring, name, make, year):
        Vehicle.__init__(self, name, make, year)
        self.sunRoof = sunRoof
        self.staring = staring
```

# Multiple Inheritance

```python
class Vehicle:
    def __init__(self, name, make, year):
        self.name = name
        self.make = make
        self.year = year
    
    def getName(self):
        return self.name

    def getMake(self):
        return self.make

    def getYear(self):
        return self.year

    # Setters
    def setName(self, name):
        self.name = name

    def setMake(self, make):
        self.make = make

    def setYear(self, year):
        self.year = year
    
        
class Car:
    def __init__(self, sunRoof, staring):
        self.sunRoof = sunRoof
        self.staring = staring
        
    # Getters
    def getSunRoof(self):
        return self.sunRoof

    def getStaring(self):
        return self.staring

    # Setters
    def setSunRoof(self, sunRoof):
        self.sunRoof = sunRoof

    def setStaring(self, staring):
        self.staring = staring

class EV:
    def __init__(self, battery):
        self.battery = battery
        
        # Getter
    def getBattery(self):
        return self.battery

    # Setter
    def setBattery(self, battery):
        self.battery = battery
        
class LuxuryVehicle:
    def __init__(self, massageChair):
        self.massageChair = massageChair
        
        # Getter
    def getMassageChair(self):
        return self.massageChair

    # Setter
    def setMassageChair(self, massageChair):
        self.massageChair = massageChair

class FbF:
    def __init__(self, fbf):
        self.fbf = fbf
        
        # Getter
    def getFbf(self):
        return self.fbf

    # Setter
    def setFbf(self, fbf):
        self.fbf = fbf
    
class LuxuryElectricFbFVehicle(Vehicle, Car, EV, LuxuryVehicle, FbF):
    def __init__(self,name, make, year, sunRoof, staring, battery, massageChair, fbf):
        Vehicle.__init__(self, name, make, year)
        Car.__init__(self, sunRoof, staring)
        EV.__init__(self, battery)
        LuxuryVehicle.__init__(self, massageChair)
        FbF.__init__(self, fbf)
    
```

```python
Etron = LuxuryElectricFbFVehicle("Etron", "Audi", 2020, "Closed", 0, 100, "off", "Not Engaged")
```
