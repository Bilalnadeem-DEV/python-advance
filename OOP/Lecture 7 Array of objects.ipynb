# Array of classes

```python
class Car:
    make = ""
    
    def __init__(self, make):
        self.make = make
        
    def displayInfo(self):
        print("Car make is: ", self.make)
        
```

```python
carsArray = []
```

```python
carMake = input("Enter make:... ")

if carMake != "":
    obj = Car(carMake)
    carsArray.append(obj)
else:
    print("Make cannot be empty")
    
```

```python
carsArray
```

```python
#display all data
for data in carsArray:
    print(data.make)
```

```python
# search for certain car
searchQuery = input("searchQuery...: ")

for data in carsArray:
    if searchQuery == data.make:
        print("Data available")
    else:
        print("No data found")
```

```python
carsArray[0].displayInfo()
```
