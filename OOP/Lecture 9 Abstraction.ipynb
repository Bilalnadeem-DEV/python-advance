```python
from abc import ABC, abstractmethod

class Car(ABC):
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    @abstractmethod
    def start_engine(self):
        pass

    @abstractmethod
    def stop_engine(self):
        pass

    @abstractmethod
    def drive(self):
        pass

class ElectricCar(Car):
    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self._battery_level = 100
        self._engine_started = False

    def stop_engine(self):
        if self._engine_started:
            self._engine_started = False
            print(f"The electric engine of {self.make} {self.model} has been stopped.")
        else:
            print("The electric engine is already off.")

    def drive(self):
        if self._engine_started:
            print(f"Driving the electric {self.make} {self.model}. Battery level: {self._battery_level}%.")
        else:
            print("You need to start the engine first.")

# Using the ElectricCar class
my_electric_car = ElectricCar("Tesla", "Model S", 2023)
my_electric_car.start_engine()  # Starts the engine
my_electric_car.drive()         # Drives the car
my_electric_car.stop_engine()   # Stops the engine
```
