```python
def generate_primes(limit):
    primes = []
    for num in range(2, limit + 1):
        for prime in primes:
            if num % prime == 0:
                break
        else:
            primes.append(num)
            yield num

# Example Usage
for prime in generate_primes(20):
    print(prime)
```

```python
class Employee:
    __id = 0

    # Constructor to initialize employee details
    def __init__(self, id):
        self.__id = id

    def _show_details(self):
        # Display employee ID (private method to be used by derived classes)
        print(f"Employee ID: {self.__id}")


class WorkDetails(Employee):
    def __init__(self, id):
        # Initialize Employee with the provided ID
        super().__init__(id)
        self.work_info = ""

    def setWorkInformation(self, val):
        # Set the work-related information for the employee
        self.work_info = val

    def getWorkDetails(self):
        # Show the work information
        self._show_details()
        print(f"Work Info: {self.work_info}")


# Example usage
work_record = WorkDetails(12345)
work_record.getWorkDetails()  # Displays ID and work info (empty at first)
work_record.setWorkInformation("Software Engineer")
work_record.getWorkDetails()  # Displays ID and updated work info
```

```python
def avengers(*avengerNames, teamStrength = 100):
    
    avengerList = []
    
    for name in avengerNames:
        avengerList.append(name)
    
    print("Avengers available: ", avengerList)
    
    if teamStrength > 150:
        print("The team is incredibly strong!")


avengers("Iron Man", "Captain America", "Thor", "Hulk", "Black Widow", 180)
avengers("Spider-Man", "Doctor Strange", teamStrength = 180)
```

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    def __init__(self, name, base_salary):
        self.name = name
        self.base_salary = base_salary

    @abstractmethod
    def calculate_bonus(self):
        pass

    @abstractmethod
    def get_role(self):
        pass


class Manager(Employee):
    def calculate_bonus(self):
        return self.base_salary * 0.3

    def get_role(self):
        return "Manager"


class Developer(Employee):
    def calculate_bonus(self):
        return self.base_salary * 0.2

    def get_role(self):
        return "Developer"


class Intern(Employee):
    def calculate_bonus(self):
        return 1000

    def get_role(self):
        return "Intern"


def display_employee_details(employee: Employee):
    print(f"Name: {employee.name}")
    print(f"Role: {employee.get_role()}")
    print(f"Base Salary: ${employee.base_salary:,.2f}")
    print(f"Bonus: ${employee.calculate_bonus():,.2f}")
    print("-" * 30)


# Example usage
employees = [
    Manager("Alice", 80000),
    Developer("Bob", 60000),
    Intern("Charlie", 15000)
]

for employee in employees:
    display_employee_details(employee)
```

```python
class Car:
    def details(self):
        return "Car: 4 seats, 200 km/h max speed."


class Truck:
    def details(self):
        return "Truck: 15 tons cargo capacity, 100 km/h max speed."


class Bike:
    def details(self):
        return "Bike: Sports type, 220 km/h max speed."


class VehicleFactory:
    @staticmethod
    def create_vehicle(vehicle_type):
        if vehicle_type == "car":
            return Car()
        elif vehicle_type == "truck":
            return Truck()
        elif vehicle_type == "bike":
            return Bike()
        else:
            raise ValueError("Unknown vehicle type!")


# Example usage
vehicles = [
    VehicleFactory.create_vehicle("car"),
    VehicleFactory.create_vehicle("truck"),
    VehicleFactory.create_vehicle("bike"),
]

for vehicle in vehicles:
    print(vehicle.details())


vehicles = [
    VehicleFactory.create_vehicle("car"),
    VehicleFactory.create_vehicle("truck"),
    VehicleFactory.create_vehicle("bike"),
]

for vehicle in vehicles:
    print(vehicle.details())
```
