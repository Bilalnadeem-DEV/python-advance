```python
class StudentWallet:
    def __init__(self, money=500):
        # Protected variable
        self._money = money

        # Private variable
        self.__secret_savings = 200

    def spend(self, amount):
        if amount > 0 and self._check_money(amount):
            self._update_money(-amount)
            print(f"Spent Rs.{amount}")
        else:
            print("Bhai, paisay nahi hain!")

    def receive(self, amount):
        self._update_money(amount)
        print(f"Received Rs.{amount}")

    def show_money(self):
        return self._money

    # Protected method
    def _update_money(self, amount):
        self._money += amount

    # Protected method
    def _check_money(self, amount):
        return self._money >= amount

    # Private method
    def __unlock_savings(self):
        self.__secret_savings += 100
        print("Secret savings unlocked! ")

    def show_savings(self):
        self.__unlock_savings()
        return self.__secret_savings


class DesperateStudent(StudentWallet):

    def ask_friend(self):
        self._update_money(300)
        print("Friend gave Rs.300. Friendship is now officially a loan.")

    def steal_from_savings(self):
        self._StudentWallet__secret_savings -= 100
        print("Student secretly stole Rs.100 from his savings.")


student = DesperateStudent(500)

student.spend(200)
print(f"Wallet: Rs.{student.show_money()}")

student.ask_friend()
print(f"Wallet: Rs.{student.show_money()}")

student.spend(700)
print(f"Wallet: Rs.{student.show_money()}")

print(f"Savings: Rs.{student.show_savings()}")

student.steal_from_savings()

print(f"Savings: Rs.{student.show_savings()}")
```
