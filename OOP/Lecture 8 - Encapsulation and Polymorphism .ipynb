```python
class Bank:
    __balance = 0
    
    def _getBalance(self):
        print("Current balance: ", self.__balance)
        
    def _setBalance(self, val):
        self.__balance = val
        
    def getBalanceTemp(self):
        print("User balance is:... ", self.__balance)
        
class Investment(Bank):
    
    def __init__(self, balance):
        self._setBalance(balance)
        
    def getUserBalance(self):
        self._getBalance()
        
    def setUSerBalance(self, val):
        self._setBalance(val)
        
    def getBalanceTemp(self):
        self._getBalance()
```

```python
balance = input("Enter balance..  ")

try:
    bal = int(balance)
    
    if bal > 100:
        # Create an object
        iObj = Investment(balance)
    else:
        print("Balance cannot be less than 100")
    
except ValueError:
    print("Error", ValueError)
```
