# Define a class CAR 

```python
class Car:
    #Car information    
    year = ""
    make = ""
    model = ""
    color = ""
    horsePower = 0
    torque = 0
    numberOfSeats = 0
    invoicePrice = 0

    #Registration information
    plateID = ""
    registeredTo = ""

    #Operation information
    isCarLocked = False
    isHeadLightOn = False
    isEngineOn = False
    currentSpeed = 0
    iswindow1Open = False
    iswindow2Open = False
    iswindow3Open = False
    iswindow4Open = False

    def __init__(self, year, make, model, color, horsePower, torque, numberOfSeats, invoicePrice):
        self.year = year
        self.make = make
        self.model = model
        self.color = color
        self.horsePower = horsePower
        self.torque = torque
        self.numberOfSeats = numberOfSeats
        self.invoicePrice = invoicePrice

    def getCompleteCarInfo(self):
        print("-------------------------------------------------------------")
        print("---------------------- CAR DETAILS --------------------------")
        print("-------------------------------------------------------------")

        print("\n <<< Car basic info >>> \n")
        print(" - Year             : ", self.year)
        print(" - Make             : ", self.make)
        print(" - Model            : ", self.model)
        print(" - Color            : ", self.color)
        print(" - Horse Power      : ", self.horsePower)
        print(" - Torque           : ", self.torque)
        print(" - Number of seats  : ", self.numberOfSeats)
        print(" - Invoice Price    : ", self.invoicePrice)

        print("\n <<< Car Registration info >>> \n")
        print(" - Plate ID         : ", self.plateID)
        print(" - Registered to    : ", self.registeredTo)

        print("\n <<< Car Operation info >>> \n")
        print(" - Car Locked?      : ", self.isCarLocked)
        print(" - Is engine on?    : ", self.isEngineOn)
        print(" - Is HeadLight on? : ", self.isHeadLightOn)
        print(" - Current speed    : ", self.currentSpeed)
        print(" - Window 1 open?   : ", self.iswindow1Open)
        print(" - Window 1 open?   : ", self.iswindow2Open)
        print(" - Window 1 open?   : ", self.iswindow3Open)
        print(" - Window 1 open?   : ", self.iswindow4Open)

    def registerCar(self, plateID, registeredTo):
        self.plateID = plateID
        self.registeredTo = registeredTo

    # Car operations
    def lockCar(self):
        self.isCarLocked = True
        print("\n\n >>>>>>> ----- Car locked")
        
    def unLockCar(self):
        self.isCarLocked = False
        print("\n\n >>>>>>> ----- Car Unlocked")
        
    def turnOnEngine(self):
        self.isEngineOn = True
        print("\n\n >>>>>>> ----- Car Engine Started")
        
    def turnOffEngine(self):
        self.isEngineOn = False
        print("\n\n >>>>>>> ----- Car Engine turned off")

    def turnOnHeadLights(self):
        self.isEngineOn = True
        print("\n\n >>>>>>> ----- Car Head Lights ON")
        
    def turnOffHeadLights(self):
        self.isEngineOn = False
        print("\n\n >>>>>>> ----- Car Head Lights OFF")
        
    def accelerateCar(self):
        self.currentSpeed = self.currentSpeed + 10
        print("\n\n >>>>>>> ----- Car speed is : ", self.currentSpeed)
        
    def applyBreaks(self):
        if self.currentSpeed >= 10:
            self.currentSpeed = self.currentSpeed - 10
        else: self.currentSpeed = 0
        print("\n\n >>>>>>> ----- Car speed is : ", self.currentSpeed)
        
    def openAllWindows(self):
        self.iswindow1Open = True
        self.iswindow2Open = True
        self.iswindow3Open = True
        self.iswindow4Open = True
        print("\n\n >>>>>>> ----- All windows opened ")
        
    def closeAllWindows(self):
        self.iswindow1Open = False
        self.iswindow2Open = False
        self.iswindow3Open = False
        self.iswindow4Open = False
        print("\n\n >>>>>>> ----- All windows opened ")
    
```
