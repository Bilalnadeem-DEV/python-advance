```python
class Car:
    name = ""
    year = "" 
    make = ""
    headLights = False
    currentSpeed = 0
    isLocked = False
    
    def __init__(self, name, year, make):
        self.name = name
        self.year = year
        self.make = make
    
    def turnOnHeadLights(self):
        self.headLights = True
    
    def turnOffHeadLights(self):
        self.headLights = False
    
    def accelerate(self):
        self.currentSpeed = self.currentSpeed + 10
    
    def decelerate(self):
        if currentSpeed > 10:
            self.currentSpeed = self.currentSpeed - 10
        else:
            self.currentSpeed = 0
    
    # Getter and setter of class
    def getName(self):
        print(self.name)
    def setName(self, name):
        self.name = name
```

```python
tucson = Car("Tucson", "2024", "Hyundai")
sportage = Car("Sportage", "2024", "Kia")
cross = Car("Cross", "2024", "Toyota")
```

```python
flag = True
cart = 0

def takeInput():
    print("<<<<<<<<<<<<<  ======   >>>>>>>>>>>>>>>>>>>>")
    print("<<<<<<<<<<<<<   Menu    >>>>>>>>>>>>>>>>>>>>")
    print("<<<<<<<<<<<<<  ======   >>>>>>>>>>>>>>>>>>>>")
    print(" ")
    print("Enter 1 to buy LCD")
    print("Enter 2 to buy Phone")
    print("Enter 3 to buy Laptop")
    print("Enter 4 to buy Fan")
    print("Enter 0 to buy Exit")
    print(" ")
    
    MenuOption = input("Enter your choice: ...  ")
    typeCast(MenuOption)
    
def typeCast(MenuOption):
    
    try:
        MenuOptionInt = int(MenuOption)
        executeMenu(MenuOptionInt)
        
    except ValueError:
        takeInput()
        
def executeMenu(choice):
    
    if choice == 1:
        print("You selected LCD")
        cart = 100
    elif choice == 2:
        print("You selected Phone")
        cart = 200

    elif choice == 3:
        print("You selected Laptop")
        cart = 300

    elif choice == 4:
        print("You selected Fan")
        cart = 400
    else:
        print("Exiting the program........")
        global flag
        flag = False
        

while flag:
    takeInput()
```

```python
class GameCharacter:
    def __init__(self, name, hp, level):
        self.name = name
        self.hp = hp
        self.level = level
        
    def show_details(self):
        print(self.name, self.hp, self.level)
        
    def run(self):
        print("character is runnin")
        
    def jump(self):
        print("character jumped")
        
    def defense(self):
        if self.hp > 10:
            self.hp = self.hp - 10
        else:
            print("Character is declared as dead")
            
    def getName(self):
        return self.name
    def getHp(self):
        return self.hp
    def getLevel(self):
        return self.level
    
    def setName(self, name):
        self.name = name
    def setHp(self, hp):
        self.hp = hp
    def setLevel(self, level):
        self.level = level    
```
