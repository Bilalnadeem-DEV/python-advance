# Simple print function polymorphic behavior

Here’s how print() adapts:

Integer: It formats and prints the integer value.

Float: It prints the floating-point number, preserving its precision.

String: It prints the string as-is.

List: It prints the list, showing all elements.

Dictionary: It prints the dictionary, showing key-value pairs.


```python
# Polymorphic behavior of print with different types

print(42)                # Printing an integer
print(3.14)              # Printing a float
print("Hello, World!")   # Printing a string
print([1, 2, 3])         # Printing a list
print({"name": "Alice"}) # Printing a dictionary
```

# Method overloading with multiple classes

```python
#Overriding

class Animal:
    def sound(self):
        return "Some generic sound"

class Dog(Animal):
    def sound(self):
        return "Bark"

class Cat(Animal):
    def sound(self):
        return "Meow"

# Polymorphic behavior:
d1 = Dog()
c1 = Cat()
a1 = Animal()


print(d1.sound())
print(c1.sound())
print(a1.sound())
```

```python
def find_max(*args):
    if args:
        return max(args)
    return None



print(find_max(10, 20, 5, 40, 25))
print(find_max(10, 20, 5, 40, 25))
print(find_max()) 
```

```python
def custom_print(*args, separator=", "):
    print(separator.join(str(arg) for arg in args))

custom_print(1, 2, 3)
custom_print("apple", "banana", "cherry", separator=" | ")
```

```python
def average_grades(*grades):
    if grades:
        return sum(grades) / len(grades)
    return 0

print(average_grades(90, 80, 85, 70))
print(average_grades()) 
```

```python
# Function to take multiple arguments
def add(datatype, *args):

    # if datatype is int
    # initialize answer as 0
    if datatype == 'int':
        answer = 0

    # if datatype is str
    # initialize answer as ''
    if datatype == 'str':
        answer = ''

    # Traverse through the arguments
    for x in args:

        # This will do addition if the
        # arguments are int. Or concatenation
        # if the arguments are str
        answer = answer + x

    print(answer)


# Integer
add('int', 5, 6)

# String
add('str', 'Hi ', 'Geeks')
```

# None as default argument 

```python
def add(a=None, b=None):
    # Checks if both parameters are available
    # if statement will be executed if only one parameter is available
    if a != None and b == None:
        print(a)
    # else will be executed if both are available and returns addition of two
    else:
        print(a+b)


# two arguments are passed, returns addition of two
add(2, 3)
# only one argument is passed, returns a
add(5502)
```

```python
class MathOperations:
    @dispatch(int, int)
    def add (self, a, b):
        print(a + b) 
    
    @dispatch(int, int, int)
    def add (self, a, b, c):
        print(a + b + c) 
    
    @dispatch(int, int, int, int)
    def add (self, a, b, c, d):
        print(a + b + c + d)

mathObj = MathOperations()
mathObj.add(10,20)
mathObj.add(10,20,30)
mathObj.add(10,20,30,40)
```

# Overriding 

```python
class Vehicle:
    def __init__(self, make, model, year, color, mileage=0):
        self.make = make
        self.model = model
        self.year = year
        self.color = color
        self.mileage = mileage

    def display_info(self):
        print(f"Vehicle Info:\nMake: {self.make}\nModel: {self.model}\nYear: {self.year}\nColor: {self.color}\nMileage: {self.mileage}")


class Car(Vehicle):
    def __init__(self, make, model, year, color, mileage, fuel_type):
        super().__init__(make, model, year, color, mileage)
        self.fuel_type = fuel_type

    # Overriding the display_info method to include fuel type
    def display_info(self):
        super().display_info()  # Call the parent method for common vehicle details
        print(f"Fuel Type: {self.fuel_type}")


# Subclass for Electric Cars
class ElectricCar(Car):
    def __init__(self, make, model, year, color, mileage, battery_capacity):
        super().__init__(make, model, year, color, mileage, fuel_type="Electric")
        self.battery_capacity = battery_capacity

    # Overriding the display_info method to include battery capacity
    def display_info(self):
        super().display_info()  # Call the parent method for common car details
        print(f"Battery Capacity: {self.battery_capacity} kWh")


# Subclass for Gasoline Cars
class GasolineCar(Car):
    def __init__(self, make, model, year, color, mileage, fuel_tank_capacity):
        super().__init__(make, model, year, color, mileage, fuel_type="Gasoline")
        self.fuel_tank_capacity = fuel_tank_capacity

    # Overriding the display_info method to include fuel tank capacity
    def display_info(self):
        super().display_info()  # Call the parent method for common car details
        print(f"Fuel Tank Capacity: {self.fuel_tank_capacity} liters")


# Create instances of ElectricCar and GasolineCar
electric_car = ElectricCar("Tesla", "Model S", 2023, "Red", 1500, battery_capacity=100)
gasoline_car = GasolineCar("Toyota", "Camry", 2022, "Blue", 12000, fuel_tank_capacity=55)

# Display information for each car
print("Electric Car Info:")
electric_car.display_info()

print("\nGasoline Car Info:")
gasoline_car.display_info()
```

```python
class Car:
    def display_info(self):
        print("This is a generic car.")

class ElectricCar(Car):
    def display_info(self):
        print("This is an electric car.")

class GasolineCar(Car):
    def display_info(self):
        print("This is a gasoline car.")

# Creating instances
generic_car = Car()
electric_car = ElectricCar()
gasoline_car = GasolineCar()

# Calling the overridden methods
generic_car.display_info()  # Output: This is a generic car.
electric_car.display_info()  # Output: This is an electric car.
gasoline_car.display_info()  # Output: This is a gasoline car.
```
